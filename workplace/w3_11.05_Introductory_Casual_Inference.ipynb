## — 잠재결과부터 빈도주의·베이지안 A/B까지

---

버튼 색을 바꿨더니 클릭이 늘었습니다. 프로모션을 보냈더니 구매가 올랐습니다. 그런데 **정말 그 변화가 만든 결과일까요?** 마침 주말이었을 수도, 원래 살 사람들이 쿠폰을 챙긴 것일 수도 있습니다.

"함께 움직인다"(상관)와 "일으켰다"(인과)는 다릅니다. 그리고 데이터로 답해야 하는 실무의 질문은 거의 언제나 **후자**입니다. 이 노트북에서는 그 질문에 답하는 가장 강력한 도구인 **무작위 실험(A/B 테스트)**을 설계하고 분석하면서, 그 바탕에 깔린 **인과추론의 언어**를 함께 익힙니다.

## 📋 학습 목표

이 노트북을 끝내면 여러분은 다음을 할 수 있습니다. 낯선 용어는 괄호에 미리 한 줄로 풀어두었습니다 — 지금 이해되지 않아도 괜찮고, 해당 Part에서 처음부터 다시 설명합니다.

1. **잠재결과(Potential Outcomes)** 프레임워크로 "인과 효과"를 정의하고, 인과추론의 근본 문제를 설명할 수 있습니다.
   - _잠재결과_ = "처치를 받았을 때의 결과"와 "받지 않았을 때의 결과", 두 개의 가능한 미래.
   - _근본 문제_ = 한 사람에게서는 그 둘 중 **하나만** 볼 수 있다는 사실.
2. **ATE·ATT**가 무엇인지 알고, 교란·선택편향이 왜 단순 비교를 망치는지 설명할 수 있습니다.
   - _ATE_ = 전체 평균 처치효과, _ATT_ = 처치받은 사람들에서의 평균 처치효과.
   - _교란_ = 처치와 결과 둘 다에 영향을 주는 숨은 제3의 변수.
3. **A/B 실험을 설계**할 수 있습니다 — 무작위 배정, 검정력·표본 크기 계산, SRM 점검.
   - _검정력_ = 진짜 효과가 있을 때 그것을 놓치지 않고 잡아낼 확률.
   - _SRM_ = 배정이 의도한 비율(예: 50:50)대로 됐는지 확인하는 점검.
4. A/B 결과를 **빈도주의(p·신뢰구간)**로 분석할 수 있습니다.
   - _빈도주의_ = "차이가 없다고 가정하면 이런 데이터가 나올 확률은?"로 묻는 익숙한 방식.
5. 같은 데이터를 **베이지안(베타 사후분포·확률적 우위)**으로도 분석하고, 두 접근을 의사결정 언어로 비교할 수 있습니다.
   - _베이지안_ = 전환율 자체를 확률분포로 보고, "B가 이길 확률 = 몇 %"를 직접 계산하는 방식.

# Part 1. 왜 인과인가 — 상관을 넘어

데이터로 묻는 질문은 세 단계로 깊어집니다. **"무엇이 일어났나"(기술)** → **"무엇과 무엇이 함께 움직이나"(상관)** → **"무엇이 무엇을 _일으키나_"(인과)**. 평균과 분포를 요약하고, 상관계수를 구하고, 회귀선을 그리는 일은 앞의 두 단계입니다. 그런데 비즈니스의 진짜 질문은 거의 항상 마지막 단계입니다.

"이 버튼 색을 바꾸면 매출이 _오를까_?", "이 신약이 회복을 _앞당길까_?", "이 정책이 고용을 _늘릴까_?" — 모두 "내가 무언가를 _바꾸면_ 결과가 어떻게 _달라지는가_"라는 인과 질문입니다.

## 상관만으로는 왜 안 되는가 — 세 가지 함정

상관계수가 크다고 인과를 말할 수 없는 이유는 크게 세 가지입니다. 이 노트북 전체가 결국 이 세 함정을 피하는 이야기입니다.

| 함정                          | 무슨 일이 벌어지나                                                              | 예                                                                       |
| ----------------------------- | ------------------------------------------------------------------------------- | ------------------------------------------------------------------------ |
| **교란(Confounding)**         | 제3의 변수가 X와 Y **둘 다**에 영향을 줘서, X와 Y가 함께 움직이는 것처럼 보인다 | 아이스크림 판매(X)와 익사 사고(Y)가 함께 늘지만, 진짜 원인은 **여름**(Z) |
| **역인과(Reverse Causation)** | 방향이 반대다. Y가 X를 일으켰는데 X가 Y를 일으킨 것으로 읽는다                  | "고객센터 문의가 많은 고객이 이탈률이 높다" → 이탈하려니까 문의한 것     |
| **우연(Chance)**              | 표본이 작거나 변수를 많이 뒤지면, 아무 관계 없는 두 변수도 우연히 함께 움직인다 | 지역별 치즈 소비량과 박사 학위 수의 상관                                 |

셋 중 가장 까다로운 것은 **교란**입니다. 역인과는 시간 순서를 따져 어느 정도 가려낼 수 있고, 우연은 표본을 늘려 줄일 수 있습니다. 그런데 교란은 — *우리가 측정하지도 못한 변수*가 범인일 수 있다는 점에서 — 아무리 데이터를 많이 모아도 저절로 사라지지 않습니다.

## 무작위 실험이 함정을 한 번에 지나가는 방법

인과를 알아내는 가장 깨끗한 방법은 **무작위 실험**입니다. 대상을 동전 던지듯 무작위로 두 집단(처치군·대조군)으로 나누면, 두 집단은 _처치 여부만 빼고_ 평균적으로 똑같아집니다.

여기서 결정적인 부분은 **"측정하지 못한 교란까지" 양쪽에 고루 섞인다**는 점입니다. 우리가 나이·성별·구매이력을 통제할 생각을 못 했더라도, 동전 던지기는 그것들을 알아서 반반으로 나눠줍니다. _알지도 못하는 변수를 통제하는 유일한 방법_ — 이것이 무작위화의 진짜 힘입니다. 그래서 결과의 차이를 *처치 덕분*이라고 말할 수 있습니다.

웹·앱 세계에서는 이 실험을 **A/B 테스트**라고 부릅니다. A안(기존)과 B안(변경)에 사용자를 무작위로 나눠 보여주고, 지표를 비교하는 것이죠.

## 이 주제가 데이터 직무에서 값진 이유

데이터 분석가·데이터 사이언티스트가 현업에서 가장 자주 마주치는 질문이 바로 "이 변화가 효과가 있었나?"입니다. 새 추천 알고리즘, 바뀐 가격 정책, 개편한 UI, 발송한 프로모션 — 회사가 무언가를 바꿀 때마다 누군가는 "그래서 그게 통했어?"를 데이터로 답해야 합니다. 그 답을 _신뢰할 수 있게_ 내놓는 능력이 곧 데이터 인력의 핵심 가치이고, 그 중심에 A/B 테스트와 인과추론이 있습니다. 실제로 빅테크 기업들은 연간 수만 건의 A/B 테스트를 돌려 제품의 방향을 정합니다 — 지금 배우는 게 바로 그 의사결정의 언어입니다.

너무 어렵게 느껴질까 봐 미리 안심시켜 드리면, 앞으로 나올 핵심 개념들은 사실 *비교*라는 단순한 발상에서 모두 흘러나옵니다. "처치받은 쪽과 안 받은 쪽을 공정하게 비교하자" — 이 한 문장을 어떻게 _제대로_ 실현하느냐가 전부입니다. 낯선 용어(잠재결과·ATE·검정력·사후분포)가 나와도 그 밑에는 늘 이 단순한 비교의 정신이 흐르고 있으니, 그것만 붙잡고 따라오시면 됩니다.

> ❓ **이 파트에서 답할 질문:** 상관으로는 왜 인과를 말할 수 없고, 무작위 실험은 그 문제를 어떻게 해결할까요?

> 🎯 **이 노트북의 핵심:** 인과는 *비교*의 문제입니다 — "처치를 받은 결과"와 "받지 않았을 결과"를 견주는 일입니다. 문제는 *한 대상에게서 둘 다 볼 수 없다*는 것(인과추론의 근본 문제)이고, **무작위 배정**이 이 문제를 _집단 수준에서_ 우회하는 마법입니다. 앞으로 내내 "비교 가능한 두 집단을 만들 수 있는가"를 물을 겁니다.

A/B 테스트의 기본 구조. 방문자를 **무작위로** 두 갈래로 나눠 A안(기존)과 B안(변경)을 각각 보여주고, 관심 지표(여기서는 클릭률)를 비교합니다. 무작위로 나눴기 때문에 두 집단은 *보여준 화면 말고는* 평균적으로 같습니다.

전체 여정을 한눈에 그리면 이렇습니다 — _잠재결과_(인과의 정의) → _ATE·ATT와 편향_(왜 단순 비교가 틀리나) → _A/B 설계_(검정력·표본·SRM) → _빈도주의 분석_(p·신뢰구간) → _베이지안 분석_(사후분포·확률적 우위) → _두 접근 비교_.

인과는 통계에서 가장 깊은 주제 중 하나지만, 한 걸음씩 쌓으면 분명히 닿습니다. 그리고 반가운 소식 하나 — **처음 만나는 통계 도구는 필요한 자리에서 그때그때 다시 설명합니다.** 검정력, 카이제곱 적합도 검정, 두 비율 z-검정, 베이즈 정리, 베타분포, 몬테카를로 — 이 여섯 가지는 통계 입문 과정에서 다루지 않거나 얕게 지나가는 것들이라, 쓰이는 시점마다 **📚 보충** 셀을 따로 두었습니다. 모르는 게 나와도 멈추지 않고 갈 수 있으니 안심하고 따라오세요.

## 인과의 사다리 — 세 단계의 질문

인과추론의 지형을 잡는 데 널리 쓰이는 틀로 Judea Pearl의 **인과의 사다리(Ladder of Causation)**가 있습니다(Pearl & Mackenzie, 2018). 방금 본 *기술 → 상관 → 인과*가 **분석 목적**의 단계였다면, 사다리는 질문이 **어떤 정보를 요구하는가**를 세 계단으로 나눕니다.

1. **연관(Association) — "본다":** "X와 Y가 함께 움직이나?" $P(Y \mid X)$. 데이터를 *관찰*하기만 하면 답할 수 있는 계단입니다.
2. **개입(Intervention) — "한다":** "내가 X를 _바꾸면_ Y가 어떻게 되나?" $P(Y \mid do(X))$ — 여기서 $do(X)$는 "X를 관찰했다"가 아니라 "X를 _내가 그렇게 만들었다_"를 뜻하는 기호입니다. **오늘의 목적지**이고, A/B 테스트가 답하는 계단입니다.
3. **반사실(Counterfactual) — "상상한다":** "*그 사람*이 그때 X를 다르게 했더라면 Y가 어땠을까?" 개인 처치효과 $\tau_i$, "나은 그 환자가 약이 없었어도 나았을까" 같은 질문이 여기 있습니다.

여기서 두 가지를 짚어두면 오해를 크게 줄일 수 있습니다.

**첫째, 계단은 *질문*의 등급이고 *방법*의 등급이 아닙니다.** "상관·회귀는 1계단"이 아닙니다. 같은 회귀라도 무엇을 조정하느냐에 따라 답하는 계단이 달라집니다 — 교란을 제대로 통제한 회귀는 2계단 질문에 답합니다 — 관찰 데이터로 인과에 다가가려는 준실험 설계(성향점수매칭·이중차분 등)가 노리는 것도 바로 그것입니다. 계단을 정하는 건 도구가 아니라 **설계와 가정**입니다.

**둘째, 잠재결과 표기를 쓴다고 3계단인 것은 아닙니다.** $Y(0), Y(1)$은 3계단까지 표현할 수 있는 *언어*지만, 그 표기를 썼다고 곧 3계단 질문을 하는 건 아닙니다. 오늘 우리가 구하는 ATE는 **2계단** 양입니다 — $E[Y(1)] - E[Y(0)] = E[Y \mid do(1)] - E[Y \mid do(0)]$ 이니까요. 정작 3계단인 개인 효과 $\tau_i$는 곧 볼 _근본 문제_ 때문에 관측할 수 없습니다. 즉 **오늘은 3계단의 언어를 빌려 2계단의 답을 구합니다.**

그런데 이 사다리는 그저 말솜씨 좋은 비유일까요? 그렇지 않습니다. 세 층이 서로 **붕괴하지 않는다**는 사실이 정리로 증명되어 있습니다(Bareinboim, Correa, Ibeling & Icard, 2022). 1계단이 완전히 똑같은데 2계단이 다른 모형 쌍이 존재하고, 2계단이 같은데 3계단이 다른 모형 쌍도 존재합니다. 그래서 관찰 데이터를 아무리 많이 모아도 개입 질문의 답은 원리적으로 결정되지 않습니다 — 데이터가 *부족해서*가 아니라 정보의 *종류*가 다르기 때문입니다. "상관은 인과가 아니다"라는 경고는 결국 *1계단에 서서 2·3계단의 답을 주장하지 말라*는 뜻이었던 것입니다.

계단을 오르려면 그에 맞는 _설계_(개입·무작위화)나 *가정*이 반드시 필요합니다. 오늘 그 사다리를 한 칸 올라봅시다.

> 📎 **표기에 관한 안내:** 사다리와 $do$ 기호는 Pearl의 **구조인과모형(SCM)** 전통에서 나왔습니다. 반면 이 노트북은 A/B 테스트에 바로 쓰이는 **잠재결과(Rubin)** 전통을 주 언어로 씁니다. 두 전통은 같은 문제를 다루는 두 표기이며, 사다리는 _우리가 지금 어디에 서 있는지_ 보여주는 지형도로만 쓰겠습니다. SCM의 인과 그래프·백도어 기준은 이 노트북에서 다루지 않습니다.
>
> 📚 **더 읽을거리:** Pearl & Mackenzie (2018), _The Book of Why_ (『왜에 관한 책』) · Pearl (2019), "The Seven Tools of Causal Inference", _CACM_ 62(3) · Bareinboim, Correa, Ibeling & Icard (2022), "On Pearl's Hierarchy and the Foundations of Causal Inference".

# Part 3. 잠재결과 프레임워크 — "인과 효과"란 정확히 무엇인가

인과를 제대로 다루려면, 먼저 "효과"가 무엇인지 _애매하지 않게_ 정의해야 합니다. 그 언어가 **잠재결과(Potential Outcomes)** 프레임워크입니다(Rubin의 인과 모형이라고도 해요).

> ❓ **이 파트에서 답할 질문:** "이 처치의 효과"를 한 사람 수준에서 어떻게 정의하고, 왜 그것을 직접 관측할 수 없을까요?

## 💡 쉽게 말하면 — 두 개의 평행우주

한 고객에게 쿠폰을 줬다고 합시다. 우리가 _정말_ 알고 싶은 건 이겁니다 — "쿠폰을 _준_ 이 고객의 구매액"과 "*같은 고객이 쿠폰을 안 받았다면*의 구매액"의 차이. 즉 **두 개의 평행우주**를 비교하는 겁니다. 쿠폰을 받은 우주에서의 결과를 $Y(1)$, 안 받은 우주에서의 결과를 $Y(0)$이라 부릅니다. 그 차이 $Y(1) - Y(0)$가 *그 사람에 대한 쿠폰의 진짜 효과*입니다. 문제는 이 "두 평행우주"를 동시에 볼 수 없다는 데 있습니다 — 우리는 쿠폰을 _주거나_ 안 주거나, 둘 중 한 우주만 살게 되니까요. 바로 다음 절에서 이 근본적인 한계를 정면으로 다룹니다.

## 정의와 '근본 문제'

각 대상 $i$는 두 개의 잠재결과를 가집니다.

- $Y_i(1)$ : 처치를 받았을 때의 결과
- $Y_i(0)$ : 처치를 안 받았을 때의 결과
- 개인 처치효과: $\tau_i = Y_i(1) - Y_i(0)$

그런데 여기에 **인과추론의 근본 문제(Fundamental Problem of Causal Inference)**가 있습니다. 우리는 한 대상에게서 _둘 중 하나만_ 관측할 수 있습니다. 쿠폰을 준 고객은 $Y(1)$만 보이고, $Y(0)$(안 줬다면)은 영원히 *반사실(Counterfactual)*로 남습니다. **개인 효과 $\tau_i$는 원리적으로 관측 불가능**합니다.

> 그래서 인과추론은 *개인*이 아니라 **집단의 평균 효과**로 눈을 돌립니다. 개인의 반사실은 못 봐도, "처치받은 집단의 평균"과 "안 받은 집단의 평균"은 비교할 수 있으니까요. 단, 그 두 집단이 *비교 가능*해야 한다는 게 핵심 조건입니다.

## 예제 · 신의 시점에서 Y(1)과 Y(0)을 모두 보기

````python
# [C2] 잠재결과 시뮬레이션 — '신의 시점'에서 Y(1)과 Y(0)을 모두 본다 (현실엔 불가능)
np.random.seed(42)                                                # 재현성: 난수를 항상 같게 고정
n = 2000
Y0 = np.random.normal(100, 20, n)                  # 처치 안 받았을 때의 구매액
true_tau = 8.0                                     # 우리가 '심은' 진짜 효과 (+8)
Y1 = Y0 + true_tau + np.random.normal(0, 5, n)     # 처치 받았을 때 (효과 + 개인차)

sim = pd.DataFrame({"Y0": Y0, "Y1": Y1})
sim["tau_i"] = sim["Y1"] - sim["Y0"]               # 개인 효과 (현실엔 못 봄)
print(sim.head().round(1))
print(f"\n[신의 시점] 진짜 ATE = E[Y(1) - Y(0)] = {sim['tau_i'].mean():.2f}  (우리가 심은 값 ≈ {true_tau})")
print("→ 현실에서는 한 사람당 Y0, Y1 중 하나만 보입니다. 이 표는 시뮬레이션이라 둘 다 보이는 것뿐이에요.")
````

> **읽는 법:** 시뮬레이션이라 한 사람의 $Y(0)$과 $Y(1)$을 _둘 다_ 적어둘 수 있습니다. 그래서 개인 효과 $\tau_i$와 그 평균(ATE)을 직접 계산했습니다. 현실에서는 각 행에서 하나의 칸이 가려진다고 상상하세요 — 그게 근본 문제입니다.

## 그림으로 · 관측되는 값과 가려지는 반사실

````python
# [C3] 근본 문제를 그림으로 — 관측되는 값과 가려지는 반사실
np.random.seed(1)                                                 # 재현성: 난수를 항상 같게 고정
idx = np.random.choice(n, 8, replace=False)
T_demo = np.random.binomial(1, 0.5, 8)             # 무작위로 처치/대조 배정
fig, ax = plt.subplots(figsize=(9, 4))                            # 그래프 하나를 그릴 도화지 준비
for k, i in enumerate(idx):
    obs = "Y1" if T_demo[k] == 1 else "Y0"
    ax.scatter(k, Y1[i], color="steelblue", s=70, alpha=1 if obs == "Y1" else 0.2)
    ax.scatter(k, Y0[i], color="indianred", s=70, alpha=1 if obs == "Y0" else 0.2)
ax.scatter([], [], color="steelblue", label="Y(1) treated")
ax.scatter([], [], color="indianred", label="Y(0) control")
ax.set_xlabel("person"); ax.set_ylabel("outcome")
ax.set_title("Observed (solid) vs counterfactual (faded) — we see only one per person")
ax.legend(); plt.tight_layout(); plt.show()                       # 그래프 간격 자동 정리 후 출력
````

## 무작위는 어떻게 반사실을 메우나 — 그리고 숨은 가정들

근본 문제 때문에 우리는 처치군의 "안 받았다면"($Y(0)$)을 볼 수 없습니다. 그런데 **무작위 배정**이 묘수를 냅니다 — 무작위로 나뉜 *대조군의 평균*이, 처치군이 *안 받았더라면의 평균*과 (평균적으로) 같아지거든요. 두 집단이 처치 빼고 똑같으니까요. 그래서 "처치군 평균 − 대조군 평균"이 진짜 ATE를 비편향 추정합니다. 즉 **대조군이 처치군의 반사실을 대신 메워주는 것**입니다.

실제 실험은 아래 흐름으로 진행됩니다. 임상시험 보고 표준(CONSORT)의 도식이지만, 웹 A/B 테스트도 구조가 같습니다 — *모집 → 무작위 배정 → 각 군의 처리 → 추적·탈락 관리 → 분석*.

<img src="https://upload.wikimedia.org/wikipedia/commons/0/00/Flowchart_of_Phases_of_Parallel_Randomized_Trial_-_Modified_from_CONSORT_2010.png" width="560" alt="Flowchart of Phases of Parallel Randomized Trial - Modified from CONSORT 2010.png">

무작위 배정 실험의 진행 흐름(CONSORT 도식). 무작위 배정 이후 각 군에서 **몇 명이 빠졌는지(탈락)** 를 반드시 기록하는 점을 보세요 — 탈락이 한쪽에 쏠리면 무작위성이 깨집니다. A/B 테스트에서 SRM 점검이 하는 역할과 같습니다.

> 🖼️ **그림 출처** — Wikimedia Commons, `Flowchart of Phases of Parallel Randomized Trial - Modified from CONSORT 2010.png` (CC BY-SA 3.0) · 원본·라이선스 상세: https://commons.wikimedia.org/wiki/File:Flowchart_of_Phases_of_Parallel_Randomized_Trial_-_Modified_from_CONSORT_2010.png

다만 이 마법에는 몇 가지 *가정*이 깔려 있습니다. 알아두면 "언제 인과 추정이 무너지는지" 보입니다.

- **교환가능성(Exchangeability):** 처치·대조군이 (처치 전엔) 교환해도 될 만큼 비슷하다. → *무작위 배정*이 이를 보장합니다. 관찰 데이터에선 깨지기 쉽죠(교란).
- **SUTVA:** ① 한 사람의 처치가 *다른 사람의 결과*에 영향을 주지 않는다(간섭 없음 — 예: 소셜네트워크·전염은 위반), ② 처치는 *한 가지 버전*이다.
- **일치성(Consistency):** 관측된 결과가 받은 처치의 잠재결과와 같다(처치가 모호하지 않다).

> 이 가정들이 그럴듯할 때만 인과 추정이 믿을 만합니다. 그래서 인과추론은 늘 "결론보다 *가정*을 먼저" 따집니다 — 이후 더 깊어질 태도입니다.

## 예제 · 대조군 평균이 처치군의 반사실을 메운다

````python
# [C4] 무작위 하에서 '대조군 평균'이 처치군의 반사실을 메운다
np.random.seed(42)                                                # 재현성: 난수를 항상 같게 고정
n = 4000
Y0 = np.random.normal(100, 20, n)
Y1 = Y0 + 8 + np.random.normal(0, 5, n)        # 진짜 효과 +8
T = np.random.binomial(1, 0.5, n)              # 무작위 배정
Y = np.where(T == 1, Y1, Y0)                   # 현실에서 관측되는 값

# '신의 시점': 처치군이 안 받았더라면의 진짜 평균
counterfactual_treated = Y0[T == 1].mean()
control_mean = Y[T == 0].mean()                # 우리가 실제로 쓰는 대용물
print(f"처치군의 반사실 E[Y(0)|T=1] (신의 시점) = {counterfactual_treated:.2f}")
print(f"대조군 평균 (현실에서 쓰는 추정치)        = {control_mean:.2f}")
print(f"→ 거의 같습니다! 무작위라 대조군이 처치군의 '안 받았더라면'을 대신합니다.")
print(f"\nATE 추정 = {Y[T==1].mean() - control_mean:.2f}  (진짜 효과 8에 근접)")
````

> **읽는 법:** "처치군이 안 받았더라면의 진짜 평균"(신의 시점에서만 아는 값)과 "대조군 평균"(현실에서 쓰는 값)이 거의 일치합니다. 무작위 배정이 바로 이 일치를 보장하기에, 우리는 못 보는 반사실을 대조군으로 _정당하게_ 대신할 수 있는 것입니다.

# Part 4. ATE·ATT와 편향 — 왜 단순 비교는 위험한가

개인 효과는 못 봐도, 우리는 _평균_ 효과를 묻습니다. 그런데 "처치군 평균 − 대조군 평균"을 그냥 계산하면 종종 _틀린_ 답이 나옵니다. 왜 그런지, 무작위 배정이 그것을 어떻게 고치는지 봅니다.

> ❓ **이 파트에서 답할 질문:** 평균 처치효과(ATE·ATT)란 무엇이고, 단순한 집단 비교는 어떤 편향에 빠질까요?

## 평균 효과의 두 얼굴 — ATE와 ATT

- **ATE (Average Treatment Effect, 평균처치효과):** *전체 모집단*에서의 평균 효과 $E[Y(1) - Y(0)]$. "모두에게 처치하면 평균 얼마나 달라지나?"
- **ATT (Average Treatment effect on the Treated, 처치군 평균효과):** *실제로 처치받은 사람들*에서의 평균 효과. "처치받은 그 사람들에게 효과가 얼마였나?"

무작위 실험에서는 처치군과 대조군이 평균적으로 같으므로 ATE ≈ ATT입니다. 하지만 _관찰 데이터_(스스로 처치를 선택한 경우)에서는 둘이 갈릴 수 있습니다 — 처치받은 사람들이 원래 특별한 집단일 수 있으니까요. 그 차이가 다음에 볼 편향의 씨앗입니다.

> 📌 **사다리와의 연결:** ATT는 "*실제로 처치받은 그 사람들*이 안 받았더라면"을 묻기 때문에, 앞서 본 사다리에서 엄밀히는 **3계단(반사실)** 양입니다 — 문헌에서 ETT(Effect of Treatment on the Treated)라고도 부릅니다. 관찰 데이터에서 ATT 추정이 왜 까다로운지가 여기서 이미 예고되는 셈이죠. 다만 *무작위 실험*에서는 처치군이 전체의 축소판이라 ATE와 ATT가 일치하므로, 오늘 A/B에서는 둘을 구분하지 않아도 됩니다.

## 💡 쉽게 말하면 — 왜 "쓴 사람 vs 안 쓴 사람"이 위험한가

"우리 앱의 프리미엄 기능을 _쓴_ 고객이 _안 쓴_ 고객보다 이탈률이 훨씬 낮다 → 프리미엄이 이탈을 막는다!" — 이 추론의 함정이 보이시나요? 프리미엄을 _스스로 선택해 쓴_ 고객은 애초에 우리 서비스에 _애착이 강한_ 사람들입니다. 그들은 프리미엄이 없었어도 덜 이탈했을 겁니다. 즉 두 집단은 _처치 여부 말고도_ 근본적으로 다릅니다. 이걸 **선택편향(Selection Bias)**, 그 배후의 숨은 차이를 **교란(Confounding)**이라 합니다.

### 📖 세 가지 구조를 구별하기 — 교란·매개·충돌

변수 세 개가 있을 때 화살표 방향에 따라 이름과 대처법이 달라집니다. 이 구별을 알면 "무엇을 통제해야 하고 무엇을 통제하면 안 되는지"가 보입니다.

<img src="https://upload.wikimedia.org/wikipedia/commons/7/70/Confounder_mediator_collider.svg" width="660" alt="Confounder mediator collider.svg">

왼쪽 **교란변수(confounder)**: 원인과 결과 양쪽에 화살표를 보냄 → *통제해야* 합니다. 가운데 **매개변수(mediator)**: 원인 → 매개 → 결과의 중간 경로 → 통제하면 효과의 일부가 사라집니다. 오른쪽 **충돌변수(collider)**: 원인과 결과가 함께 영향을 주는 변수 → 통제하면 **없던 상관이 생깁니다**.

> 🖼️ **그림 출처** — Wikimedia Commons, `Confounder mediator collider.svg` (CC BY-SA 4.0) · 원본·라이선스 상세: https://commons.wikimedia.org/wiki/File:Confounder_mediator_collider.svg

| 구조 | 그림 | 통제해야 하나 | 통제하면 |
| --- | --- | --- | --- |
| **교란** | X ← Z → Y | ✅ 해야 함 | 편향이 줄어든다 |
| **매개** | X → M → Y | ❌ (총효과를 원하면) | 효과의 일부가 지워진다 |
| **충돌** | X → C ← Y | ❌ 절대 | *가짜 상관*이 새로 생긴다 |

> 💡 D015에서 언급한 **"나쁜 통제(bad control)"** 가 바로 매개·충돌변수를 통제하는 경우입니다. "변수를 많이 넣으면 안전하다"는 흔한 오해가 여기서 깨집니다 — **무엇을 넣을지는 데이터가 아니라 인과 구조에 대한 지식이 정합니다.**

## 시뮬레이션 · 선택이 만든 가짜 효과 vs 무작위

````python
# [C6] 편향 시뮬레이션 — '스스로 선택'이 만든 가짜 효과 vs 무작위가 준 진짜 효과
np.random.seed(42)                                                # 재현성: 난수를 항상 같게 고정
n = 4000
X = np.random.normal(0, 1, n)                       # 숨은 교란 (예: 서비스 애착도)
true_effect = 5.0                                   # 우리가 심은 진짜 처치효과

# (A) 관찰 데이터: 애착(X) 높은 사람이 처치(프리미엄)를 더 선택
p_self = 1 / (1 + np.exp(-2 * X))
T_obs = np.random.binomial(1, p_self)                             # 이항 난수 — n번 시도 중 성공 횟수
Y_obs = 50 + true_effect * T_obs + 12 * X + np.random.normal(0, 5, n)   # X도 결과를 끌어올림
naive = Y_obs[T_obs == 1].mean() - Y_obs[T_obs == 0].mean()

# (B) 무작위 실험: 동전 던지듯 배정 (X와 무관)
T_rand = np.random.binomial(1, 0.5, n)                            # 이항 난수 — n번 시도 중 성공 횟수
Y_rand = 50 + true_effect * T_rand + 12 * X + np.random.normal(0, 5, n)
rand = Y_rand[T_rand == 1].mean() - Y_rand[T_rand == 0].mean()

print(f"진짜 효과(심음)            : {true_effect:.1f}")
print(f"(A) 관찰 데이터 단순 비교  : {naive:5.1f}   ← 편향! (교란 X가 부풀림)")
print(f"(B) 무작위 실험 단순 비교  : {rand:5.1f}   ← 진짜 효과에 근접")
````

> **읽는 법:** 똑같이 "처치군 평균 − 대조군 평균"을 계산했는데, (A) 관찰 데이터에서는 진짜 효과(5)보다 한참 부풀려진 값이 나옵니다. 애착 높은 사람이 처치를 더 선택했고, 그들은 원래 결과도 높았으니까요(교란). 반면 (B) 무작위 배정에서는 교란 X가 양쪽에 고루 섞여, 단순 비교만으로도 진짜 효과에 근접합니다. **무작위화가 편향을 녹이는 장면**입니다.

## 그림으로 · 배정별 교란 X의 분포 비교

# Part 5. A/B 실험 설계 — 제대로 설계하기

무작위 배정만으로 충분하지 않습니다. _표본이 너무 작으면_ 진짜 효과가 있어도 못 잡고, _배정이 의도대로 안 됐으면_ 결과 자체가 오염됩니다. 그래서 실험은 _시작하기 전에_ 설계해야 합니다.

> ❓ **이 파트에서 답할 질문:** 신뢰할 만한 A/B 결과를 얻으려면, 실험 전에 무엇을(표본 크기·검정력·배정 점검) 정해야 할까요?

## 💡 쉽게 말하면 — 실험은 '사전 계획'이 9할

A/B 테스트의 흔한 실수는 "일단 돌리고 유의해지면 멈추기"입니다. 이건 뒤에서 자세히 볼 **'훔쳐보기(peeking)'**로, 거짓 양성을 잔뜩 만듭니다. 제대로 된 실험은 마치 *과학 실험계획서*처럼, 시작 전에 세 가지를 못 박습니다 — **(1) 얼마나 작은 차이를 잡고 싶은가(최소 효과), (2) 그걸 잡을 확률을 얼마로 보장할까(검정력), (3) 그러려면 표본이 몇 명 필요한가.** 이 계산을 **검정력 분석(Power Analysis)**이라고 합니다.

### 📚 보충 — 두 종류의 오류와 '검정력'

가설검정에서 유의수준 $\alpha$는 익숙할 겁니다. **검정력**은 그 짝인데 입문 과정에서 자주 생략됩니다. A/B 설계의 핵심이라 여기서 확실히 잡고 갑시다.

실험의 결론은 네 칸으로 정리됩니다. 세로는 우리가 **내린 판정**, 가로는 **숨겨진 진실**입니다.

|                             | **진짜로 효과가 있다**                         | **진짜로 효과가 없다**                       |
| --------------------------- | ---------------------------------------------- | -------------------------------------------- |
| 검정이 **"효과 있다"** 판정 | 🎯 올바른 발견 — 확률 $1-\beta$ = **검정력**   | ❌ **1종 오류** — 확률 $\alpha$, 헛것을 봤다 |
| 검정이 **"효과 없다"** 판정 | ❌ **2종 오류** — 확률 $\beta$, 있는 걸 놓쳤다 | 🎯 올바른 판단 — 확률 $1-\alpha$             |

- **$\alpha$ (1종 오류율):** 효과가 _없는데_ "있다"고 잘못 외칠 확률. 관례적으로 0.05.
- **$\beta$ (2종 오류율):** 효과가 _있는데_ "없다"고 놓칠 확률.
- **검정력 $1-\beta$:** 효과가 있을 때 **놓치지 않고 잡아낼** 확률. 관례적으로 0.80을 목표로 합니다.

A/B 테스트에서 두 오류의 **비용은 다릅니다.** 1종 오류는 *효과 없는 기능을 출시*하게 만들고(개발·운영 비용 낭비), 2종 오류는 _좋은 기능을 버리게_ 만듭니다(기회 손실). 어느 쪽이 더 아픈지는 상황에 따라 다르므로, α와 검정력을 **실험 전에** 정해두는 것이 중요합니다.

#### 네 숫자가 서로 묶여 있습니다

$$\alpha \;\longleftrightarrow\; \text{검정력}\,(1-\beta) \;\longleftrightarrow\; \text{잡고 싶은 효과 크기} \;\longleftrightarrow\; \text{표본 크기}\; n$$

**넷 중 셋을 정하면 남은 하나가 자동으로 결정됩니다.** 실무에서는 α = 0.05, 검정력 = 0.80, "잡고 싶은 최소 효과"를 먼저 정하고 → **필요한 $n$을 역산**합니다. 이 역산이 **검정력 분석(Power Analysis)**이고, 바로 아래에서 계산해봅니다.

> ⚠️ **꼭 기억할 직관:** 잡고 싶은 효과가 *절반*으로 작아지면 필요한 표본은 대략 *네 배*로 늘어납니다 — 표본 크기가 효과 크기의 제곱에 반비례($n \propto 1/\text{효과}^2$)하기 때문입니다. "작은 개선을 확인하려면 훨씬 큰 실험이 필요하다"는 뜻이고, A/B 테스트가 종종 수만 명을 요구하는 이유가 바로 이것입니다.

> 🔎 **최소 검출 효과(MDE, Minimum Detectable Effect):** 표본이 이미 정해져 있을 때(예: 하루 트래픽이 1만 명뿐) 거꾸로 계산하는 값입니다. "이 표본으로는 최소 몇 %p 이상의 효과만 볼 수 있다"를 알려주죠. MDE가 우리가 기대하는 개선폭보다 크다면 — **그 실험은 시작해도 답을 얻을 수 없습니다.** 실험을 아예 하지 않기로 결정하는 것도 검정력 분석의 중요한 쓸모입니다.

## 검정력과 표본 크기

위 보충에서 정리한 네 숫자의 관계를 실제로 계산해봅니다.

- **검정력(Power, $1-\beta$):** 진짜 효과가 있을 때 그것을 잡아낼 확률. 보통 0.8(80%)을 목표로 합니다.
- **표본 크기:** 잡고 싶은 효과가 _작을수록_, 원하는 검정력이 _높을수록_ 더 많은 표본이 필요합니다.

전환율을 비교하는 A/B라면 두 가지를 정하면 됩니다.

1. **기준 전환율** — 현재 A안의 전환율 (예: 10%)
2. **잡고 싶은 최소 향상폭** — 이만큼은 꼭 잡아내고 싶다 (예: +2%p, 즉 12%로)

그러면 필요한 표본을 역산할 수 있습니다. 아래 코드는 `statsmodels`의 `NormalIndPower`를 씁니다 — 두 비율을 `proportion_effectsize`로 표준화된 효과 크기($h$)로 바꾼 뒤, α·검정력과 함께 넣어 $n$을 구하는 구조입니다.

````python
# [C8] 검정력 분석 — 전환율 10% → 12%(=+2%p)를 80% 검정력으로 잡으려면?
base, target = 0.10, 0.12
es = proportion_effectsize(target, base)        # 두 비율의 효과 크기
n_per_group = NormalIndPower().solve_power(effect_size=es, alpha=0.05, power=0.8, alternative="two-sided")  # α·검정력·효과 크기 → 필요한 표본 수를 역산
print(f"효과 크기(h) = {es:.4f}")
print(f"필요한 표본 = 그룹당 약 {np.ceil(n_per_group):,.0f}명  (총 {np.ceil(n_per_group)*2:,.0f}명)")

# 잡고 싶은 향상폭이 작아질수록 표본이 폭증한다
print("\n[향상폭별 그룹당 필요 표본]")
for tgt in [0.105, 0.11, 0.12, 0.14]:
    e = proportion_effectsize(tgt, base)                          # 두 비율을 표준화된 효과 크기 h로 변환
    nn = NormalIndPower().solve_power(effect_size=e, alpha=0.05, power=0.8)  # α·검정력·효과 크기 → 필요한 표본 수를 역산
    print(f"  10% → {tgt*100:.1f}%  (+{(tgt-base)*100:.1f}%p) : 약 {np.ceil(nn):>8,.0f}명")
````

> **읽는 법:** 잡고 싶은 향상폭이 +0.5%p처럼 작아지면 그룹당 수만~수십만 명이 필요합니다. "표본을 충분히 모을 수 있는가"가 곧 "이 실험이 현실적인가"를 결정합니다. 이 계산을 _실험 전에_ 해두면, "며칠을 돌려야 하는지"가 나옵니다.

## 배정 점검 — SRM (Sample Ratio Mismatch)

50:50으로 배정했다면 실제 두 그룹의 크기도 _대략_ 반반이어야 합니다. 만약 한쪽이 유독 많거나 적으면(예: 52:48이 우연이라기엔 너무 치우침), 배정 로직·로깅에 버그가 있다는 신호 — 이걸 **SRM**이라 합니다. SRM이 있으면 _결과를 믿기 전에 파이프라인부터 고쳐야_ 합니다. 점검은 간단합니다. 카이제곱 적합도 검정으로 "관측 비율이 기대 50:50과 다른가"를 봅니다.

### 📚 보충 — 카이제곱 적합도 검정 (SRM 점검에 쓰는 도구)

SRM 점검은 "관측된 배정 인원수가 **의도한 비율**에서 벗어났는가"를 묻습니다. 이럴 때 쓰는 것이 **카이제곱 적합도 검정(Chi-square Goodness-of-Fit)**입니다.

발상은 단순합니다 — **관측값이 기대값에서 얼마나 벗어났는지**를 모든 칸에 대해 더합니다.

$$\chi^2 = \sum_{i} \frac{(O_i - E_i)^2}{E_i}$$

- $O_i$ = 실제로 관측된 개수, $E_i$ = 의도대로였다면 기대되는 개수
- 각 칸의 차이를 **제곱**해서(부호를 없애고) **기대값으로 나눈 뒤**(규모를 보정) 모두 더합니다.
- 이 값이 **클수록** "의도한 비율과 다르다"는 증거가 강해집니다.

예를 들어 10,000명을 50:50으로 배정하려 했는데 5,100:4,900이 나왔다면, 기대값은 각 5,000명이므로

$$\chi^2 = \frac{(5100-5000)^2}{5000} + \frac{(4900-5000)^2}{5000} = 2 + 2 = 4$$

이 $\chi^2$ 값을 자유도 1의 카이제곱분포와 견주어 p값을 얻습니다(칸이 2개면 자유도 = 2 − 1 = 1).

#### 해석 방향이 평소와 반대입니다 ⚠️

보통 가설검정에서는 "p가 작으면 좋은 발견"이었습니다. 그런데 SRM 점검에서 우리가 _바라는_ 결과는 **"차이가 없다"**입니다.

| p값                      | 뜻                                      | 판단                                   |
| ------------------------ | --------------------------------------- | -------------------------------------- |
| **크다** (예: 0.01 이상) | 의도한 비율과 다르다고 볼 근거가 약하다 | ✅ 배정 정상 — 분석을 진행             |
| **작다** (예: 0.01 미만) | 의도한 비율에서 유의하게 벗어났다       | 🚨 **SRM 발생** — 분석 중단, 원인 조사 |

> ⚠️ **SRM이 뜨면 결과를 해석하지 말고 파이프라인을 의심하세요.** 배정 버그, 봇 트래픽, 로깅 누락, 리다이렉트 실패 등이 흔한 원인입니다. 이 경우 두 집단은 이미 "비교 가능"하지 않으므로, 그 위에서 계산한 p값이나 사후분포는 아무 의미가 없습니다.
>
> 💡 SRM 점검은 보통 $\alpha$를 0.05보다 **훨씬 작게**(0.01 또는 0.001) 잡습니다. 정상적인 실험을 괜히 중단시키는 일을 줄이기 위해서입니다. 아래 코드는 0.01을 기준으로 판정합니다.

## 확인 · SRM 점검 (카이제곱 적합도)

````python
# [C9] SRM 점검 — 배정이 의도(50:50)대로 됐나? (카이제곱 적합도)
n_a, n_b = 9980, 10020          # 실제 배정된 두 그룹 크기 (정상 예시)
total = n_a + n_b
chi2_srm, p_srm = stats.chisquare([n_a, n_b], [total/2, total/2])  # 카이제곱 적합도 검정 — 관측이 기대 비율과 맞나
print(f"그룹 크기: A={n_a}, B={n_b}")
print(f"SRM 검정: chi2 = {chi2_srm:.2f}, p = {p_srm:.3f}")
print("→", "정상 (p가 크면 50:50과 유의차 없음)" if p_srm > 0.01 else "⚠️ SRM 의심! 배정/로깅 점검 필요")

# 만약 7000:13000처럼 치우쳤다면?
bad = stats.chisquare([7000, 13000], [10000, 10000])              # 카이제곱 적합도 검정 — 관측이 기대 비율과 맞나
print(f"\n[치우친 예 7000:13000] p = {bad.pvalue:.2e} → ⚠️ 심각한 SRM, 결과 분석 보류")
````

> **읽는 법:** 정상 배정은 p가 커서 "50:50과 다르지 않음"으로 나옵니다. 반면 7000:13000처럼 치우치면 p가 사실상 0 — _우연으로 보기 어려운_ 불일치이고, 분석을 멈추고 배정 시스템을 점검해야 합니다. SRM 점검은 A/B 결과를 믿기 위한 *첫 관문*입니다.

> 📌 **다른 산업에서는?** 실험 설계의 검정력·표본 계산은 임상시험(필요 환자 수 산정·IRB 승인), 제조 품질실험, 정책 파일럿 설계의 표준 절차입니다. SRM 점검은 대규모 온라인 실험 플랫폼(빅테크 A/B 시스템)의 필수 가드레일입니다.

## 가장 비싼 실수 — '훔쳐보기(Peeking)'의 위험

A/B 테스트에서 초보자가 가장 자주 저지르는 치명적 실수가 있습니다. **"실험을 돌리면서 매일 결과를 들여다보다가, p가 0.05 밑으로 떨어지는 순간 '유의하다!'며 멈추는 것"**입니다. 직관적으로는 합리적 같지만, 통계적으로는 1종 오류(거짓 양성)를 *폭발*시킵니다. 검정을 여러 번 반복하면 "한 번이라도 우연히 유의해질" 확률이 누적되기 때문이에요(다중비교의 그 원리!). 시뮬레이션으로 직접 확인해봅시다.

````python
# [C10] peeking 시뮬 — 진짜 효과가 '0'인데도, 매일 들여다보면 거짓 양성이 폭증
np.random.seed(3)                                                 # 재현성: 난수를 항상 같게 고정
runs = 1500
fp_peek = 0      # 중간에 한 번이라도 유의하면 '유의'로 카운트
fp_once = 0      # 끝에 딱 한 번만 검정
for _ in range(runs):
    cA = cB = nA = nB = 0
    hit = False
    for step in range(10):                      # 10번 들여다봄
        cA += np.random.binomial(400, 0.10); nA += 400            # 이항 난수 — n번 시도 중 성공 횟수
        cB += np.random.binomial(400, 0.10); nB += 400      # A=B, 진짜 효과 0
        if proportions_ztest([cB, cA], [nB, nA])[1] < 0.05:       # 두 비율 z-검정 — 비율 차이를 표준오차로 나눔
            hit = True
    if hit: fp_peek += 1
    if proportions_ztest([cB, cA], [nB, nA])[1] < 0.05: fp_once += 1  # 두 비율 z-검정 — 비율 차이를 표준오차로 나눔
print(f"진짜 효과 = 0 인데도...")
print(f"  끝에 한 번만 검정: 거짓 양성률 = {fp_once/runs:.1%}   (의도한 5% 근처 ✓)")
print(f"  10번 훔쳐보기    : 거짓 양성률 = {fp_peek/runs:.1%}   (5%를 훨씬 초과! ✗)")
````

> **읽는 법:** A와 B가 _완전히 같은데도_, 중간에 여러 번 들여다보며 "유의해지면 멈추기"를 하면 거짓 양성률이 의도한 5%를 크게 넘습니다. 그래서 실험은 **표본 크기를 미리 정하고 그때까지 결과를 보지 않는 것**이 원칙입니다. (꼭 중간 점검이 필요하면 _순차검정·알파 소비_ 같은 보정 기법을 씁니다.) 이것이 Part 5에서 "표본을 사전에 계산"한 진짜 이유입니다.

> ➕ **더 알고 싶다면 — MDE:** 표본 크기를 정하면, 그 표본으로 "80% 검정력으로 잡을 수 있는 _가장 작은 효과_"가 정해집니다. 이를 **MDE(Minimum Detectable Effect, 최소검출효과)**라 해요. "우리 실험은 +1%p까지는 잡지만 +0.3%p는 못 잡는다"처럼, 실험의 *해상도*를 미리 아는 겁니다.

# Part 6. 빈도주의 A/B — p와 신뢰구간

이제 실험이 끝났다고 합시다. A안과 B안의 전환 결과(클릭/구매 여부)를 받았습니다. 가장 전통적인 분석이 **빈도주의(Frequentist)** 접근 — 지금까지 배운 가설검정과 신뢰구간을 그대로 씁니다.

> ❓ **이 파트에서 답할 질문:** A와 B의 전환율 차이가 "우연으로 보기 어려운가"를 p값과 신뢰구간으로 어떻게 판정할까요?

## 💡 쉽게 말하면 — 빈도주의가 던지는 질문

빈도주의의 질문은 이렇습니다 — **"두 안에 진짜 차이가 *없다*고 가정했을 때(귀무가설), 지금 관측한 만큼의 차이가 우연히 나올 확률(p)은?"** p가 충분히 작으면 "우연이라기엔 너무 큰 차이"라며 귀무가설을 기각하고 "차이가 있다"고 봅니다.

여기서 한 가지만 새롭습니다. 지금까지 평균을 비교할 때는 t-검정을 썼을 겁니다. 그런데 A/B에서 우리가 비교하는 것은 "전환했다/안 했다"의 **비율**입니다. 비율에 맞춘 도구가 **두 비율 z-검정(Two-proportion z-test)**입니다 — 바로 아래 보충에서 설명합니다.

### 📚 보충 — 두 비율 z-검정은 무엇을 하는가

t-검정은 **평균**을 비교합니다. 그런데 A/B의 결과는 보통 "전환했다(1)/안 했다(0)"이고, 우리가 비교하는 것은 **비율**입니다. 비율에 맞춘 검정이 **두 비율 z-검정(Two-proportion z-test)**입니다.

**아이디어는 t-검정과 완전히 같습니다** — "차이"를 "차이가 자연스럽게 흔들리는 폭"으로 나눕니다.

$$z = \frac{\hat p_B - \hat p_A}{SE}, \qquad SE = \sqrt{\;\hat p\,(1-\hat p)\left(\frac{1}{n_A} + \frac{1}{n_B}\right)}$$

- $\hat p_A, \hat p_B$ = 각 안에서 관측된 전환율 (전환수 ÷ 인원수)
- $\hat p$ = 두 집단을 **합쳐서** 계산한 전환율. 귀무가설이 "둘이 같다"이므로, 그 가정 아래에서의 공통 비율을 씁니다.
- 분모 $SE$(표준오차) = "진짜 차이가 0일 때 관측 차이가 우연히 흔들리는 폭"

즉 $z$가 뜻하는 것은 **"관측된 차이가 우연의 흔들림 몇 배인가"**입니다. $|z|$가 2 근처를 넘어서면 p값이 0.05 아래로 내려갑니다.

**왜 t가 아니고 z인가?** 비율은 0/1 값의 평균이라, 표본이 충분히 크면 그 차이의 분포가 정규분포에 가까워집니다(중심극한정리). 그래서 t분포 대신 표준정규분포 $z$를 씁니다. 실무 기준으로 **각 집단의 전환수와 비전환수가 모두 10 이상**이면 이 근사는 안전합니다 — A/B 테스트는 보통 수천 명 규모라 문제가 되는 일이 드뭅니다.

> 💡 **카이제곱 검정과의 관계:** 2×2 표(안 × 전환여부)에 카이제곱 독립성 검정을 하면 **수학적으로 같은 결론**이 나옵니다($\chi^2 = z^2$). 그래도 A/B에서 z-검정을 선호하는 이유는, z-검정이 차이의 **방향**(어느 쪽이 높은지)과 **신뢰구간**을 바로 주기 때문입니다.

#### 신뢰구간을 반드시 함께 봐야 합니다

p값은 "차이가 있나 없나"만 답합니다. 정작 의사결정에 필요한 건 **"얼마나 차이 나나"**죠. 그래서 차이의 95% 신뢰구간을 같이 구합니다.

$$(\hat p_B - \hat p_A) \;\pm\; 1.96 \times SE$$

읽는 법은 두 가지입니다.

1. **구간이 0을 포함하는가** — 포함하지 않으면 "유의하다"는 뜻이고, p < 0.05와 대응합니다.
2. **구간의 폭이 얼마인가** — 우리 추정이 얼마나 정밀한지를 보여줍니다.

두 번째가 특히 중요합니다. 아래 두 결과는 점추정이 똑같이 +2.4%p인데도 결정이 완전히 달라집니다.

| 신뢰구간           | 읽는 법                         | 결정                 |
| ------------------ | ------------------------------- | -------------------- |
| `[+0.8%p, +4.0%p]` | 최소 0.8%p는 오른 것으로 보인다 | 출시 검토 가능       |
| `[-0.1%p, +4.9%p]` | 떨어졌을 가능성도 남아 있다     | 판단 보류, 표본 추가 |

> ⚠️ 검정에는 합친 비율 $\hat p$로 만든 표준오차를 쓰지만, 신뢰구간에는 두 집단을 **따로** 추정한 표준오차를 쓰는 것이 표준입니다. 귀무가설을 가정하지 않고 차이 자체를 추정하는 것이 목적이니까요. 아래 코드의 `confint_proportions_2indep`이 그 계산을 해줍니다.

## 예제 · 실험 결과 집계와 배정 점검

````python
# [C12] 실험 결과 집계 — 실제 무작위 배정 현장실험(ResumeNames)
# 그룹 A = afam형 이름, 그룹 B = cauc형 이름 (이름은 무작위 배정됨)
grp_A, grp_B = "afam", "cauc"
n_A = int((resume["ethnicity"] == grp_A).sum())
n_B = int((resume["ethnicity"] == grp_B).sum())
conv_A = int(((resume["ethnicity"] == grp_A) & (resume["call"] == "yes")).sum())
conv_B = int(((resume["ethnicity"] == grp_B) & (resume["call"] == "yes")).sum())
rate_A, rate_B = conv_A / n_A, conv_B / n_B

print(f"A({grp_A}형 이름): {conv_A}/{n_A} = {rate_A:.3%}")
print(f"B({grp_B}형 이름): {conv_B}/{n_B} = {rate_B:.3%}")
print(f"관측된 연락률 차이: {(rate_B - rate_A)*100:+.2f}%p")

# 실험 위생 점검 — 배정 비율이 의도대로였나(SRM)
_, p_srm0 = stats.chisquare([n_A, n_B], [(n_A + n_B) / 2] * 2)    # 카이제곱 적합도 검정 — 관측이 기대 비율과 맞나
print(f"\n[배정 점검] A:B = {n_A}:{n_B},  SRM 검정 p = {p_srm0:.3f}"
      f" → {'정상 (50:50과 유의차 없음)' if p_srm0 > 0.01 else '⚠️ SRM 의심'}")
````

## 예제 · 두 비율 z-검정 + 차이의 95% 신뢰구간

````python
# [C13] 빈도주의 검정 — 두 비율 z-검정 + 차이의 95% 신뢰구간
z, p = proportions_ztest([conv_B, conv_A], [n_B, n_A])     # B-A 순서
ci_low, ci_high = confint_proportions_2indep(conv_B, n_B, conv_A, n_A, method="wald")  # 두 비율 '차이'의 신뢰구간 (0을 포함하면 비유의)
print(f"비율 z-검정 : z = {z:.3f}, p = {p:.4f}")
print(f"전환율 차이 (B - A) 95% CI : [{ci_low*100:+.2f}%p, {ci_high*100:+.2f}%p]")
print()
if p < 0.05:
    print("→ p < 0.05: '차이가 없다'를 기각. B의 전환율이 통계적으로 더 높다고 판단.")
else:
    print("→ p >= 0.05: 차이가 유의하지 않음(우연 배제 못 함).")
````

> **읽는 법:** p가 0.05보다 훨씬 작아 "A와 B가 같다"는 귀무가설을 기각합니다. 신뢰구간 `[+1.7%p, +4.7%p]`이 **0을 포함하지 않으므로** 같은 결론입니다. 게다가 신뢰구간은 "차이가 대략 어디서 어디 사이"라는 *크기 정보*까지 줍니다 — p 하나보다 훨씬 풍부합니다(앞서 배운 교훈!).
>
> 💡 **무작위 배정이라 해석이 달라집니다.** D014에서 같은 데이터를 카이제곱으로 분석했을 때는 "연관이 있다"까지만 말했습니다. 그런데 이름이 **무작위로 배정**됐다는 사실을 알고 나면, 이 차이를 *배정된 요인의 효과*로 해석할 근거가 생깁니다. **같은 숫자라도 '어떻게 데이터가 만들어졌는지'가 해석의 등급을 결정합니다** — 이것이 오늘 노트북 전체의 요지입니다.

## 그림으로 · 두 전환율과 95% 신뢰구간

````python
# [C14] [도식] 두 전환율과 95% 신뢰구간 비교
from statsmodels.stats.proportion import proportion_confint
ci_A = proportion_confint(conv_A, n_A, method="wilson")           # 한 비율의 신뢰구간 (Wilson 방법이 소표본에 안전)
ci_B = proportion_confint(conv_B, n_B, method="wilson")           # 한 비율의 신뢰구간 (Wilson 방법이 소표본에 안전)
fig, ax = plt.subplots(figsize=(7, 4))                            # 그래프 하나를 그릴 도화지 준비
ax.errorbar([rate_A, rate_B], [0, 1],                             # 오차막대(신뢰구간) 표시
            xerr=[[rate_A-ci_A[0], rate_B-ci_B[0]], [ci_A[1]-rate_A, ci_B[1]-rate_B]],
            fmt="o", capsize=6, color="steelblue")
ax.set_yticks([0, 1]); ax.set_yticklabels(["A (기존)", "B (신규)"])
ax.set_xlabel("conversion rate"); ax.set_title("A vs B conversion rate (95% CI)")
plt.tight_layout(); plt.show()                                    # 그래프 간격 자동 정리 후 출력
````

> **읽는 법(그림):** 두 점이 각 그룹의 연락률, 가로선이 95% 신뢰구간입니다. 두 구간이 **겹치지 않아** 차이가 뚜렷하다는 시각 신호입니다. 다만 빈도주의의 한계도 기억하세요 — p값은 "B가 A보다 나을 확률"을 _직접_ 알려주지 않습니다. 그게 바로 다음 Part 베이지안이 답하는 질문입니다.

> 📌 **다른 산업에서는?** 비율 비교 A/B는 이메일 개봉률, 광고 CTR, 가입 전환, 약물 반응률(이분형) 등 "예/아니오 결과"를 비교하는 모든 실험의 기본 분석입니다.

## 차이를 말하는 세 가지 방법 — 그리고 실질 유의성

"B가 A보다 낫다"를 숫자로 말하는 방법은 하나가 아닙니다. 같은 결과라도 표현에 따라 *체감*이 크게 달라지니, 셋을 모두 알아두면 보고가 정직해집니다.

- **절대 위험차(Absolute Difference):** $p_B - p_A$. "전환율이 1.8%p 올랐다." 가장 정직한 표현.
- **상대 위험(Relative Risk, RR):** $p_B / p_A$. "전환율이 1.16배(16%) 올랐다." 커 보이게 만들기 쉬움.
- **오즈비(Odds Ratio, or):** 오즈의 비. 로지스틱 회귀·의학에서 표준.

> ⚠️ "**전환율 16% 상승!**"(상대)과 "**1.8%p 상승**"(절대)은 *같은 결과*입니다. 마케팅 자료는 큰 쪽(상대)을 즐겨 쓰지만, 의사결정에는 절대차와 *기준값*을 함께 봐야 속지 않습니다.

## 예제 · 같은 결과를 세 가지로 표현하기

````python
# [C15] 같은 A/B 결과를 세 가지로 표현 (Part 6의 conv/n 재사용)
rd = rate_B - rate_A                                            # 절대 위험차
rr = rate_B / rate_A                                           # 상대 위험
orr = (conv_B*(n_A-conv_A)) / (conv_A*(n_B-conv_B))            # 오즈비
print(f"기준 전환율 A = {rate_A:.2%}")
print(f"절대 위험차  = {rd*100:+.2f}%p     ('정직한' 크기)")
print(f"상대 위험 RR = {rr:.3f}  (= +{(rr-1)*100:.1f}%, 커 보이는 표현)")
print(f"오즈비   OR  = {orr:.3f}")
````

## 확인 · 표본이 크면 사소한 차이도 유의해진다

````python
# [C16] 실질 유의성의 함정 — 표본이 크면 '아무 의미 없는' 차이도 유의해진다
print("표본 크기별: 전환율 10.0% vs 10.2% (고작 +0.2%p)의 p값")
for n in [1_000, 10_000, 100_000, 1_000_000]:
    cA, cB = int(n*0.100), int(n*0.102)
    pv = proportions_ztest([cB, cA], [n, n])[1]                   # 두 비율 z-검정 — 비율 차이를 표준오차로 나눔
    flag = "유의(p<0.05)" if pv < 0.05 else "비유의"
    print(f"  n={n:>9,} : p = {pv:.4f}  → {flag}")
print("\n→ 표본이 100만이면 +0.2%p도 '유의'합니다. 하지만 비즈니스적으로 의미 있을까요?")
````

> **읽는 법:** 표본이 커질수록 _지극히 작은_ 차이도 통계적으로 "유의"해집니다. 그래서 **통계적 유의성(p)과 실질적 유의성(효과 크기·비즈니스 가치)을 반드시 구분**해야 해요. "유의하다"는 "우연이 아니다"일 뿐, "중요하다"가 아닙니다. 절대 위험차와 비용·편익을 함께 봐야 진짜 의사결정이 됩니다 — 내내 반복된 그 교훈입니다.

> 📌 **다른 산업에서는?** 상대 위험·오즈비는 의학(치료 효과), 절대 위험차는 공중보건(정책 효과 규모), 효과 크기는 마케팅 ROI 판단에서 핵심 지표입니다.

# Part 7. 베이지안 A/B — "B가 이길 확률"을 직접 구하기

빈도주의 p값은 유용하지만, 의사결정자가 진짜 묻고 싶은 건 따로 있습니다 — **"그래서 B가 A보다 나을 확률이 몇 %인가요?"** 빈도주의는 이 질문에 직접 답하지 못해요(p는 "차이가 없다는 가정 하의 데이터 확률"이지, "B가 낫다는 확률"이 아니니까요). **베이지안 A/B**는 바로 이 질문에 답합니다.

> ❓ **이 파트에서 답할 질문:** "B가 A보다 나을 확률"을 어떻게 직접 계산하고, 그것을 어떻게 의사결정에 쓸까요?

### 📚 보충 — 베이즈 정리, 3분 요약

베이지안 A/B를 이해하려면 **베이즈 정리** 하나만 알면 됩니다. 한 문장으로는 이렇습니다.

> **원래 믿고 있던 것**을, **새로 본 데이터**로 **갱신한다.**

수식은 이렇게 생겼습니다.

$$\underbrace{P(\theta \mid \text{데이터})}_{\text{사후}} \;\propto\; \underbrace{P(\text{데이터} \mid \theta)}_{\text{우도}} \;\times\; \underbrace{P(\theta)}_{\text{사전}}$$

세 단어만 익히면 끝입니다.

| 용어                     | 기호                           | 뜻                                            | A/B에서는                                            |
| ------------------------ | ------------------------------ | --------------------------------------------- | ---------------------------------------------------- |
| **사전분포** (prior)     | $P(\theta)$                    | 데이터를 보기 *전*의 믿음                     | "B의 전환율이 얼마쯤일지 아직 모른다"                |
| **우도** (likelihood)    | $P(\text{데이터} \mid \theta)$ | 전환율이 $\theta$라면 이 데이터가 나올 가능성 | "전환율이 12%라면 1000명 중 120명 전환은 그럴듯하다" |
| **사후분포** (posterior) | $P(\theta \mid \text{데이터})$ | 데이터를 본 *뒤*의 갱신된 믿음                | "B의 전환율은 11~13% 사이일 것 같다"                 |

$\propto$ 기호는 "비례한다"는 뜻입니다. 전체 합이 1이 되도록 맞추는 상수를 생략했다는 표시인데, 실용적으로는 신경 쓸 필요가 없습니다.

#### 익숙한 예로 감을 잡아봅시다

어떤 질병의 유병률이 1%이고, 검사의 정확도가 99%라고 합시다. 검사에서 **양성**이 나왔다면 실제로 병에 걸렸을 확률은 얼마일까요?

직관은 "99%"라고 답하지만, 실제로는 **약 50%**입니다. 이유는 사전분포 때문입니다 — 애초에 건강한 사람이 99배 많으니, 1%의 오진율만으로도 "건강한데 양성"인 사람이 "병 있고 양성"인 사람과 비슷한 수로 생깁니다.

**교훈:** 데이터(검사 결과)만 보면 안 되고, **사전 정보(유병률)와 결합**해야 올바른 결론이 나옵니다. 베이즈 정리는 이 결합을 정확하게 해주는 규칙입니다.

#### 이것이 A/B와 무슨 상관인가

빈도주의는 "차이가 없다고 가정할 때 이 데이터가 나올 확률"(p값)을 계산했습니다. 질문의 방향이 **데이터 쪽**이죠. 그런데 의사결정자가 진짜 알고 싶은 것은 반대 방향입니다.

```
빈도주의 :  P(이런 데이터 | 차이가 없다)        ← p값
베이지안 :  P(B가 A보다 낫다 | 관측한 데이터)    ← 우리가 원했던 답
```

베이즈 정리는 이 방향을 **뒤집을 수 있게** 해줍니다. 대신 대가가 하나 있습니다 — 사전분포를 정해야 한다는 것이죠. 이 대가와 이득을 Part 8에서 나란히 비교합니다.

> 💡 사전분포가 주관적이라는 점이 베이지안 비판의 핵심이었습니다. 실무적 답은 두 가지입니다. ① 정보가 없을 때는 **평평한 사전분포**를 써서 데이터가 결론을 지배하게 한다. ② 사전분포를 바꿔가며 결론이 얼마나 흔들리는지 **민감도 분석**을 한다. 이 Part 뒤쪽 [C21]에서 실제로 확인합니다.

## 💡 쉽게 말하면 — 믿음을 데이터로 업데이트

위 보충에서 본 베이즈 정리를 A/B에 그대로 적용해봅시다. 베이지안 A/B는 각 안의 *전환율*에 대한 믿음을 **확률분포**로 표현하고, 데이터를 보며 그 분포를 업데이트합니다. 전환율은 0과 1 사이의 비율이므로, 여기에 딱 맞는 분포가 필요합니다. 그게 **베타분포(Beta)**입니다.

### 베타분포 — 비율에 대한 믿음을 담는 그릇

베타분포 $\text{Beta}(a, b)$는 **0과 1 사이에서만** 정의되는 분포입니다. 모수가 두 개인데, 직관적으로 이렇게 읽으면 됩니다.

$$\text{Beta}(a,\; b) \quad\longleftrightarrow\quad \text{"성공을 } a-1 \text{번, 실패를 } b-1 \text{번 본 것 같은 믿음"}$$

- $\text{Beta}(1, 1)$ = 0~1 어디든 똑같이 가능 → **완전히 평평한 분포**(균등분포와 같습니다). "아무 정보 없음"을 뜻해서 출발점으로 자주 씁니다.
- $\text{Beta}(11, 91)$ = 성공 10번, 실패 90번을 본 믿음 → 0.1 근처에 봉우리가 생기고, 폭도 적당히 좁습니다.
- $\text{Beta}(101, 901)$ = 성공 100번, 실패 900번 → 봉우리는 여전히 0.1 근처인데 훨씬 **뾰족**합니다. 같은 비율이라도 데이터가 많을수록 확신이 커진다는 뜻이죠.

### 갱신 공식이 놀랍도록 간단합니다

실험 전에는 "전환율이 0~1 어디든 가능"이라는 평평한 믿음 $\text{Beta}(1,1)$에서 시작합니다. 그리고 전환 성공 $s$번, 실패 $f$번을 관측하면 사후분포는 그냥

$$\text{사후분포} = \text{Beta}(1 + s,\; 1 + f)$$

가 됩니다. 적분도, 복잡한 계산도 필요 없습니다. **관측한 성공·실패 횟수를 모수에 더하기만 하면 끝**입니다.

이렇게 깔끔한 이유는 베타분포와 이항 데이터(성공/실패)가 **켤레(conjugate)** 관계이기 때문입니다. 켤레란 "사전분포가 베타면 사후분포도 다시 베타가 된다"는 성질을 말합니다. 분포의 *종류*는 그대로고 _모수만_ 바뀌니, 데이터가 들어올 때마다 숫자 두 개만 갱신하면 됩니다. A/B 테스트에서 베이지안이 실용적인 큰 이유가 이것입니다.

### 관점의 전환 — 확률에 확률을 매긴다

처음 베이지안을 접하면 이 발상이 낯설 수 있습니다. 빈도주의에서 전환율은 *고정된 미지의 숫자*였습니다 — 우리가 모를 뿐 참값은 하나로 정해져 있다고 봅니다. 반면 베이지안은 전환율 자체를 *불확실한 양*으로 보고, 그 불확실성을 분포로 표현합니다.

"전환율이 정확히 얼마다"가 아니라 "11% 근처일 가능성이 높고 9%나 13%일 가능성은 낮다"처럼 *믿음의 분포*로 말하는 것입니다. 데이터가 쌓일수록 이 분포는 점점 좁고 뾰족해집니다 — 더 확신하게 되는 겁니다.

> 💡 **왜 이 전환이 실무에서 유용한가:** 전환율이 분포라면, "B의 전환율이 A보다 클 확률"도 **분포에서 곧바로 계산**할 수 있습니다. 빈도주의의 p값로는 답할 수 없었던 "그래서 B가 이길 확률이 몇 %냐?"라는 질문에 직접 답할 수 있게 되는 것이죠. 이걸 어떻게 구하는지가 이 Part의 하이라이트입니다.

베이지안 A/B 분석의 전형적인 산출물. 두 안의 전환율에 대한 **사후분포**를 그리고, 그 겹침의 정도로 "어느 쪽이 더 나은가"와 "얼마나 확실한가"를 함께 읽습니다. 빈도주의가 p값 하나를 주는 자리에, 베이지안은 분포 전체를 줍니다.

> 아래에서 우리 실험 데이터로 이 그림을 직접 그리고, 거기서 **P(B > A)** 와 **기대손실**까지 뽑아냅니다.

## 예제 · 베타 사후분포 그리기

````python
# [C18] 베타 사후분포 — 데이터를 본 뒤 각 안의 전환율에 대한 믿음
# (Part 6의 conv_A, conv_B, n_A, n_B 재사용)
a_post_A, b_post_A = 1 + conv_A, 1 + (n_A - conv_A)    # Beta(1+성공, 1+실패)
a_post_B, b_post_B = 1 + conv_B, 1 + (n_B - conv_B)

xs = np.linspace(0.09, 0.15, 500)                                 # 구간을 균등 간격으로 쪼갠 x축 값
fig, ax = plt.subplots(figsize=(8, 4))                            # 그래프 하나를 그릴 도화지 준비
ax.plot(xs, stats.beta.pdf(xs, a_post_A, b_post_A), color="indianred", label="A posterior")  # 베타분포의 확률밀도 — 비율에 대한 믿음의 모양
ax.plot(xs, stats.beta.pdf(xs, a_post_B, b_post_B), color="steelblue", label="B posterior")  # 베타분포의 확률밀도 — 비율에 대한 믿음의 모양
ax.fill_between(xs, stats.beta.pdf(xs, a_post_A, b_post_A), alpha=0.2, color="indianred")  # 베타분포의 확률밀도 — 비율에 대한 믿음의 모양
ax.fill_between(xs, stats.beta.pdf(xs, a_post_B, b_post_B), alpha=0.2, color="steelblue")  # 베타분포의 확률밀도 — 비율에 대한 믿음의 모양
ax.set_xlabel("conversion rate"); ax.set_ylabel("density")
ax.set_title("Posterior distributions of conversion rate"); ax.legend()
plt.tight_layout(); plt.show()                                    # 그래프 간격 자동 정리 후 출력
````

> **읽는 법(그림):** 두 개의 종 모양이 각각 A·B의 전환율에 대한 *사후 믿음*입니다. 봉우리는 관측된 전환율, 폭은 불확실성(표본이 많을수록 좁아짐)입니다. B(파랑)의 분포가 A(빨강)보다 오른쪽에 있지만 *겹치는 부분*도 있습니다 — 그 겹침이 "B가 확실히 낫다고 단정 못 하는" 불확실성입니다.

### 📚 보충 — 몬테카를로: 수식이 어려우면 그냥 뽑아보면 된다

이제 "B가 A보다 나을 확률" $P(p_B > p_A)$를 구해야 합니다. 두 베타분포의 차이에 대한 확률이라 손으로 적분하면 꽤 번거롭습니다. 그런데 아주 쉬운 우회로가 있습니다.

**몬테카를로(Monte Carlo)** 방법의 발상은 이렇습니다.

> 확률을 *계산*하기 어렵다면, **아주 많이 뽑아보고 세어보면 된다.**

절차는 세 줄입니다.

1. A의 사후분포에서 전환율 하나를 뽑는다 → $p_A^{(1)}$
2. B의 사후분포에서 전환율 하나를 뽑는다 → $p_B^{(1)}$
3. $p_B^{(1)} > p_A^{(1)}$ 인지 확인한다 → 참/거짓

이걸 20만 번 반복해서 **참이었던 비율**을 세면, 그게 곧 $P(B > A)$입니다. 20만 번 중 19만 2천 번 참이었다면 $P(B>A) \approx 96\%$인 것이죠.

```text
뽑기 1 : p_A=0.104, p_B=0.118  →  B 승  ✓
뽑기 2 : p_A=0.109, p_B=0.115  →  B 승  ✓
뽑기 3 : p_A=0.112, p_B=0.108  →  A 승  ✗
   ...  (20만 번 반복)
                  → B가 이긴 비율 = P(B > A)
```

**왜 이게 맞는가?** 큰 수의 법칙 때문입니다. 표본을 많이 뽑을수록 "뽑힌 비율"은 "진짜 확률"에 수렴합니다. 동전을 20만 번 던지면 앞면 비율이 0.5에 매우 가까워지는 것과 같은 원리입니다.

**덤으로 얻는 것들이 많습니다.** 뽑아둔 표본이 있으면 확률 하나만이 아니라 여러 질문에 바로 답할 수 있습니다.

| 궁금한 것          | 뽑은 표본으로 계산하는 방법                     |
| ------------------ | ----------------------------------------------- |
| B가 이길 확률      | `(draws_B > draws_A).mean()`                    |
| 기대 향상폭        | `(draws_B - draws_A).mean()`                    |
| 향상폭의 95% 구간  | `np.percentile(draws_B - draws_A, [2.5, 97.5])` |
| 3%p 이상 오를 확률 | `((draws_B - draws_A) > 0.03).mean()`           |

수식을 새로 유도할 필요 없이 **질문을 코드 한 줄로 바꿔 쓰면 되는 것** — 이것이 베이지안이 실무에서 강력한 이유입니다.

> ⚠️ **주의 두 가지:** ① 뽑는 횟수가 적으면 결과가 흔들립니다. 보통 수만~수십만 번을 씁니다. ② 난수를 쓰므로 실행마다 값이 미세하게 달라집니다. 재현이 필요하면 `np.random.seed()`로 고정하세요 — 아래 코드도 그렇게 합니다.
>
> 💡 여기서 나오는 **신용구간(credible interval)**은 신뢰구간(confidence interval)과 이름이 비슷하지만 뜻이 다릅니다. 신용구간은 "이 구간에 참값이 있을 확률이 95%"라고 **곧이곧대로** 읽을 수 있습니다. 신뢰구간은 그렇게 읽으면 틀리죠("같은 실험을 무한히 반복하면 그 중 95%의 구간이 참값을 덮는다"가 정확한 뜻입니다). 베이지안이 직관적이라고 하는 이유 중 하나입니다.

## 예제 · P(B > A)를 몬테카를로로 계산하기

````python
# [C19] "B가 A보다 나을 확률" — 두 사후분포에서 표본을 뽑아 직접 계산 (Monte Carlo)
np.random.seed(0)                                                 # 재현성: 난수를 항상 같게 고정
draws_A = np.random.beta(a_post_A, b_post_A, 200000)              # 베타 사후분포에서 표본 추출 (몬테카를로)
draws_B = np.random.beta(a_post_B, b_post_B, 200000)              # 베타 사후분포에서 표본 추출 (몬테카를로)
p_B_better = (draws_B > draws_A).mean()
lift = draws_B - draws_A
print(f"P(B > A)              = {p_B_better:.1%}   ← 의사결정자가 진짜 원하던 답!")
print(f"기대 향상폭 E[B - A]  = {lift.mean()*100:+.2f}%p")
print(f"향상폭 95% 신용구간   = [{np.percentile(lift,2.5)*100:+.2f}%p, {np.percentile(lift,97.5)*100:+.2f}%p]")  # 표본에서 분위수 계산 (신용구간 양 끝)
print(f"손해 볼 확률 P(B < A) = {(1-p_B_better):.1%}  (B로 바꿨을 때 오히려 나쁠 위험)")
````

> **읽는 법:** 베이지안은 **"B가 A보다 나을 확률 = OO%"**라는, 의사결정자가 곧장 쓸 수 있는 답을 줍니다. "95% 신용구간(Credible Interval)"은 빈도주의 신뢰구간과 달리 *"참값이 이 구간에 있을 확률이 95%"*라고 (드물게) 직접 해석할 수 있습니다. 또 "B로 바꿨을 때 손해 볼 확률"까지 보여줘, *위험을 감안한 의사결정*에 자연스럽습니다.

> 📌 **다른 산업에서는?** 베이지안 A/B는 온라인 실험 플랫폼(작은 표본에서도 의사결정, 조기 종료 규칙), 추천·랭킹의 멀티암 밴딧, 의료 적응형 임상시험 등에서 점점 표준이 되고 있습니다.

## "이길 확률"을 넘어 — 기대손실로 결정하기

P(B>A)가 92%라고 해서 무조건 B를 출시해야 할까요? 더 똑똑한 의사결정 도구가 **기대손실(Expected Loss)**입니다. "만약 B를 골랐는데 _사실 A가 더 나았다면_ 평균적으로 얼마나 손해 보는가"를 사후분포로 계산하는 것입니다. 이 기대손실이 *우리가 감내할 수 있는 작은 임계값*보다 낮으면 "B를 출시해도 안전하다"고 판단합니다.

## 예제 · 기대손실로 결정하기

````python
# [C20] 기대손실(expected loss) — 베이지안 의사결정 규칙
# (Part 7의 draws_A, draws_B 또는 a_post 재사용; 안전하게 재계산)
np.random.seed(0)                                                 # 재현성: 난수를 항상 같게 고정
dA = np.random.beta(1+conv_A, 1+n_A-conv_A, 300000)               # 베타 사후분포에서 표본 추출 (몬테카를로)
dB = np.random.beta(1+conv_B, 1+n_B-conv_B, 300000)               # 베타 사후분포에서 표본 추출 (몬테카를로)
loss_choose_B = np.maximum(dA - dB, 0).mean()    # B 골랐는데 A가 나을 때의 평균 손해
loss_choose_A = np.maximum(dB - dA, 0).mean()    # A 골랐는데 B가 나을 때의 평균 손해
print(f"P(B>A)               = {(dB>dA).mean():.1%}")
print(f"B 선택 시 기대손실    = {loss_choose_B*100:.4f}%p   (작을수록 B가 안전)")
print(f"A 선택 시 기대손실    = {loss_choose_A*100:.4f}%p")
tol = 0.001   # 감내 가능한 손실 임계 (0.1%p)
print(f"\n임계 {tol*100:.2f}%p 기준 →", "B 출시 (기대손실이 임계 미만)" if loss_choose_B < tol else "추가 데이터 수집 권장")
````

> **읽는 법:** "B를 골랐을 때의 기대손실"이 우리가 정한 작은 임계값보다 낮으면, 설령 100% 확신은 아니어도 "잘못돼봐야 손해가 미미하니 출시해도 된다"는 합리적 결정이 됩니다. P(B>A)라는 *확률*에 손실의 *크기*까지 결합한, 한 단계 성숙한 의사결정 규칙입니다.

## 확인 · 사전분포가 결론을 얼마나 바꾸나

````python
# [C21] 사전분포(prior)의 영향 — 데이터가 적을 때 특히 중요
# 실험 데이터에서 그룹당 100건만 뽑아, 표본이 작은 상황을 재현합니다.
tiny = resume.groupby("quality", group_keys=False).sample(100, random_state=1)  # 집단별로 묶어서 계산
small_nA = int((tiny["quality"] == "low").sum())
small_nB = int((tiny["quality"] == "high").sum())
small_cA = int(((tiny["quality"] == "low")  & (tiny["call"] == "yes")).sum())
small_cB = int(((tiny["quality"] == "high") & (tiny["call"] == "yes")).sum())
print(f"작은 표본: A {small_cA}/{small_nA},  B {small_cB}/{small_nB}\n")

for a0, b0, label in [(1, 1, "약한 사전 Beta(1,1)"),
                      ("info", "info", "정보적 사전 Beta(8,92)≈8%")]:
    if a0 == "info":
        a0, b0 = 8, 92          # "과거 실험에서 연락률은 대략 8%였다"는 사전 지식
    np.random.seed(0)                                             # 재현성: 난수를 항상 같게 고정
    dA_p = np.random.beta(a0 + small_cA, b0 + small_nA - small_cA, 200000)  # 베타 사후분포에서 표본 추출 (몬테카를로)
    dB_p = np.random.beta(a0 + small_cB, b0 + small_nB - small_cB, 200000)  # 베타 사후분포에서 표본 추출 (몬테카를로)
    print(f"{label:28s}: P(B>A) = {(dB_p > dA_p).mean():.1%},"
          f"  기대 향상 = {(dB_p - dA_p).mean()*100:+.2f}%p")
print("\n→ 표본이 작으면 사전분포가 결론에 꽤 영향을 줍니다(많아지면 데이터가 사전을 압도).")
````

> **읽는 법:** 그룹당 100건뿐인 작은 표본에서는 어떤 사전분포를 쓰느냐가 P(B>A)를 눈에 띄게 바꿉니다. *약한(평평한) 사전*은 데이터가 결론을 지배하게 두고, *정보적 사전*("과거 연락률은 대략 8%")은 추정을 그 값 쪽으로 끌어당깁니다.
>
> 그래서 베이지안 분석은 **사전분포를 반드시 공개해야** 합니다. 실무 원칙은 두 가지입니다 — ① 정보가 없으면 **약한 사전**을 써서 데이터가 말하게 한다, ② 사전을 바꿔가며 결론이 얼마나 흔들리는지 **민감도 분석**을 하고 그 결과를 함께 보고한다. 표본이 충분히 커지면(앞의 2,400건씩) 두 사전의 결론이 거의 같아집니다.

| 의사결정         | "유의/비유의" 이분법 경향                         | 확률·위험 기반(연속적)                   |
| 강점             | 표준적·규제 친화·간단                             | 직관적 해석·소표본·조기 의사결정         |

> 💡 보통 두 결론은 *일치*합니다 — 데이터가 충분하면 p가 작을 때 P(B>A)도 높습니다. 차이는 *답의 언어*에 있습니다. "유의하다"(빈도주의)는 게 곧 "비즈니스적으로 도입할 만하다"는 뜻은 아닙니다. 그래서 효과 크기·위험·비용과 함께 봐야 합니다.

## 예제 · 같은 데이터, 두 렌즈의 답을 나란히

````python
# [C23] 같은 데이터, 두 렌즈의 답을 나란히
print("=" * 52)
print("같은 A/B 데이터에 대한 두 접근의 답")
print("=" * 52)
print(f"[빈도주의]  p = {p:.4f},  차이 95% CI = [{ci_low*100:+.2f}, {ci_high*100:+.2f}]%p")
print(f"            → 결론: {'유의 (차이 있음)' if p < 0.05 else '비유의'}")
print(f"[베이지안]  P(B>A) = {p_B_better:.1%},  기대 향상폭 = {lift.mean()*100:+.2f}%p")
print(f"            → 결론: B가 나을 확률이 {p_B_better:.0%}, 손해 볼 위험 {1-p_B_better:.0%}")
print("=" * 52)
print("두 접근 모두 'B가 더 낫다'를 가리키지만, 빈도주의는 '유의/비유의', 베이지안은 '확률·위험'으로 말합니다.")
````

> **읽는 법:** 두 접근이 같은 방향을 가리킵니다. 이 실험은 효과가 뚜렷해서 빈도주의는 "p < 0.001로 유의", 베이지안은 "B가 나을 확률이 사실상 100%, 손해 볼 위험 0%"라고 말합니다 — **증거가 충분할 때 두 렌즈는 다투지 않습니다.**
>
> 경영진에게는 후자의 언어가 더 와닿는 경우가 많습니다 — "유의수준 0.05" 대신 "B로 바꿔서 손해 볼 확률은 사실상 0입니다"라고 말할 수 있으니까요.
>
> 💡 **그런데 진짜 차이는 결과가 애매할 때 드러납니다.** 종합 실습에서 같은 실험의 다른 요인(이력서 품질)을 분석하면, 빈도주의는 "유의하지 않음(p = 0.07)", 베이지안은 "B가 나을 확률 96%"라고 답합니다. **같은 데이터, 다른 언어** — 그 상황을 어떻게 결정으로 옮기는지가 종합 실습의 핵심입니다.

## 데이터가 쌓이면 두 렌즈는 한 점으로 모인다

빈도주의와 베이지안이 "다른 답을 준다"고 오해하기 쉽지만, 실은 _충분한 데이터_ 앞에서 두 결론은 같은 곳으로 수렴합니다. 진짜 차이가 있다면, 표본이 커질수록 빈도주의 p는 0으로, 베이지안 P(B>A)는 1로 _함께_ 향합니다. 확인해봅시다.

````python
# [C24] 표본이 커질수록 p↓ 와 P(B>A)↑ 가 나란히 수렴
# 참 전환율은 이력서 실험에서 관측된 두 비율(7.3% vs 8.7%)을 그대로 씁니다.
np.random.seed(11)                                                # 재현성: 난수를 항상 같게 고정
p_true_A, p_true_B = 0.0734, 0.0875
print(f"가정한 참 비율: A = {p_true_A:.3f}, B = {p_true_B:.3f}  (관측된 실험 결과와 동일)")
print(f"\n{'n/그룹':>10} | {'빈도주의 p':>12} | {'베이지안 P(B>A)':>16}")
print("-" * 46)
for n in [500, 2500, 10000, 40000]:
    cA_s = np.random.binomial(n, p_true_A)                        # 이항 난수 — n번 시도 중 성공 횟수
    cB_s = np.random.binomial(n, p_true_B)                        # 이항 난수 — n번 시도 중 성공 횟수
    pv = proportions_ztest([cB_s, cA_s], [n, n])[1]               # 두 비율 z-검정 — 비율 차이를 표준오차로 나눔
    dA_s = np.random.beta(1 + cA_s, 1 + n - cA_s, 100000)         # 베타 사후분포에서 표본 추출 (몬테카를로)
    dB_s = np.random.beta(1 + cB_s, 1 + n - cB_s, 100000)         # 베타 사후분포에서 표본 추출 (몬테카를로)
    print(f"{n:>10,} | {pv:>12.4f} | {(dB_s > dA_s).mean():>15.1%}")
````

> **읽는 법:** 표에서 두 가지를 보세요.
>
> 1. **표본이 작을 때(n=500)는 두 지표가 애매할 뿐 아니라, 아예 *엉뚱한 방향*을 가리킬 수도 있습니다.** 참 비율은 B가 더 높은데도 우연히 A가 더 나은 표본이 뽑혀 P(B>A)가 50%를 밑돌기도 합니다. 이것이 "표본이 작으면 실험 결과를 믿지 말라"의 실제 모습입니다.
> 2. 반면 데이터가 쌓이면 빈도주의(p↓)와 베이지안(P(B>A)↑)이 _나란히_ 확신을 더해가며 같은 결론으로 수렴합니다.
>
> 둘은 경쟁자가 아니라, 같은 진실을 다른 언어로 말하는 동료입니다. 그래서 표본이 충분하면 **"어느 방법을 쓰느냐"보다 "데이터가 충분한가"가 훨씬 중요합니다.**

## 언제 무엇을 쓸까

- **빈도주의:** 규제·논문·표준 리포팅이 필요할 때, 조직이 익숙할 때, 사전 정보가 마땅찮을 때.
- **베이지안:** 표본이 작거나 빨리 의사결정해야 할 때, "이길 확률·손실 위험"을 직접 말해야 할 때, 과거 실험의 사전 지식을 넣고 싶을 때.

> 둘 다 _가정과 결론을 분리해 정직하게_ 보고하는 게 핵심입니다. p나 P(B>A) 같은 *숫자 하나*로 결정을 자동화하지 말고, 효과 크기·비용·위험과 함께 *판단*하세요. 통계는 의사결정을 _돕는_ 것이지 _대신하는_ 게 아닙니다.

실무에서는 사실 둘을 _함께_ 쓰는 경우가 많습니다. 빈도주의로 "유의한가"를 표준적으로 보고하면서, 베이지안으로 "그래서 출시했을 때 이길 확률과 손해 위험은 얼마인가"를 의사결정자에게 전달하는 식입니다. 한 분석에 두 렌즈를 나란히 담으면, 통계에 익숙한 사람과 비즈니스 언어를 쓰는 사람 모두에게 설득력을 갖습니다. 중요한 건 *방법론 진영 다툼*이 아니라, 각 도구가 답하는 질문이 무엇인지 알고 상황에 맞게 고르는 일입니다. "내 질문에 답해주는 도구가 무엇인가"를 먼저 묻는 습관이, 인과 분석에서도 그대로 통합니다.

마지막으로 강조하고 싶은 건, **어떤 통계 방법도 나쁜 실험 설계를 구해주지 못한다**는 점입니다. 무작위 배정이 깨졌거나(SRM), 표본이 턱없이 부족하거나, peeking으로 데이터를 오염시켰다면 — 빈도주의든 베이지안이든 그 위에서 내린 결론은 모래 위의 성입니다. 그래서 인과 분석의 진짜 승부처는 *분석 기법*이 아니라 **설계와 가정**에 있습니다. "이 두 집단이 정말 비교 가능한가?"라는 질문에 자신 있게 "그렇다"고 답할 수 있을 때, 비로소 어떤 분석을 얹어도 신뢰할 수 있습니다.

# 🧪 종합 실습 — "애매한 결과"를 어떻게 결정할까

같은 이력서 실험에는 무작위로 배정된 요인이 하나 더 있습니다 — **이력서 품질**(`quality`: `low` / `high`). 경력·자격을 보강한 이력서(B = high)가 연락률을 높였을까요?

이 실습이 중요한 이유: **결과가 깔끔하지 않습니다.** 실무 A/B의 대부분이 이렇습니다 — p는 0.05를 살짝 넘고, 신뢰구간은 0을 걸치고, 그래도 결정은 내려야 합니다. 그때 빈도주의와 베이지안이 각각 무엇을 말해주는지 나란히 보겠습니다.

```text
[1] 설계 점검 — 배정 비율(SRM)
[2] 빈도주의 — z-검정·신뢰구간
[3] 베이지안 — 사후분포·P(B>A)·기대손실
[4] 의사결정 보고서 — "채택할까?" (가정·결론 분리)
```

## 종합 [1] · 이력서 품질 요인 집계와 SRM 점검

````python
# [C25] 종합 실습 데이터 — 같은 실험의 또 다른 무작위 요인(이력서 품질)
qA_label, qB_label = "low", "high"          # A = 기존(품질 낮음), B = 개선(품질 높음)
nA = int((resume["quality"] == qA_label).sum())
nB = int((resume["quality"] == qB_label).sum())
cA = int(((resume["quality"] == qA_label) & (resume["call"] == "yes")).sum())
cB = int(((resume["quality"] == qB_label) & (resume["call"] == "yes")).sum())

print(f"A(품질 {qA_label}) : {cA}/{nA} = {cA/nA:.3%}")
print(f"B(품질 {qB_label}): {cB}/{nB} = {cB/nB:.3%}")
print(f"관측된 차이: {(cB/nB - cA/nA)*100:+.2f}%p")

# [1] SRM 점검 — 무작위 배정이 의도대로 됐나
_, p_srm = stats.chisquare([nA, nB], [(nA + nB) / 2] * 2)         # 카이제곱 적합도 검정 — 관측이 기대 비율과 맞나
print(f"\n[1] SRM: A:B = {nA}:{nB},  p = {p_srm:.3f} →",
      "배정 정상" if p_srm > 0.01 else "⚠️ SRM — 분석 보류")
````

## 종합 [2][3] · 빈도주의와 베이지안 한 번에

````python
# [C26] [2] 빈도주의 + [3] 베이지안 한 번에
z, pv = proportions_ztest([cB, cA], [nB, nA])                     # 두 비율 z-검정 — 비율 차이를 표준오차로 나눔
lo, hi = confint_proportions_2indep(cB, nB, cA, nA, method="wald")  # 두 비율 '차이'의 신뢰구간 (0을 포함하면 비유의)
print(f"[2] 빈도주의: z={z:.2f}, p={pv:.4f}, 차이95%CI=[{lo*100:+.2f},{hi*100:+.2f}]%p")

np.random.seed(0)                                                 # 재현성: 난수를 항상 같게 고정
dA = np.random.beta(1+cA, 1+nA-cA, 200000)                        # 베타 사후분포에서 표본 추출 (몬테카를로)
dB = np.random.beta(1+cB, 1+nB-cB, 200000)                        # 베타 사후분포에서 표본 추출 (몬테카를로)
pBA = (dB > dA).mean(); liftS = (dB - dA)
print(f"[3] 베이지안: P(B>A)={pBA:.1%}, 기대향상={liftS.mean()*100:+.2f}%p, 손실위험={1-pBA:.1%}")
````

## 종합 · 사후분포와 향상폭 분포 시각화

````python
# [C27] 종합 — 두 안의 사후분포와 향상폭 분포 시각화
np.random.seed(0)                                                 # 재현성: 난수를 항상 같게 고정
dA = np.random.beta(1 + cA, 1 + nA - cA, 200000)                  # 베타 사후분포에서 표본 추출 (몬테카를로)
dB = np.random.beta(1 + cB, 1 + nB - cB, 200000)                  # 베타 사후분포에서 표본 추출 (몬테카를로)

fig, axes = plt.subplots(1, 2, figsize=(12, 3.8))                 # 그림 한 장에 그래프 2개(1행 2열)
xs = np.linspace(0.05, 0.12, 400)                                 # 구간을 균등 간격으로 쪼갠 x축 값
axes[0].plot(xs, stats.beta.pdf(xs, 1 + cA, 1 + nA - cA), color="indianred", label="A (low)")  # 베타분포의 확률밀도 — 비율에 대한 믿음의 모양
axes[0].plot(xs, stats.beta.pdf(xs, 1 + cB, 1 + nB - cB), color="steelblue", label="B (high)")  # 베타분포의 확률밀도 — 비율에 대한 믿음의 모양
axes[0].fill_between(xs, stats.beta.pdf(xs, 1 + cA, 1 + nA - cA), alpha=0.2, color="indianred")  # 베타분포의 확률밀도 — 비율에 대한 믿음의 모양
axes[0].fill_between(xs, stats.beta.pdf(xs, 1 + cB, 1 + nB - cB), alpha=0.2, color="steelblue")  # 베타분포의 확률밀도 — 비율에 대한 믿음의 모양
axes[0].set_title("Posteriors of callback rate"); axes[0].legend(); axes[0].set_xlabel("rate")

lift_draws = (dB - dA) * 100
axes[1].hist(lift_draws, bins=60, color="seagreen", alpha=0.7)
axes[1].axvline(0, color="black", ls="--", label="차이 = 0")        # 세로 기준선 표시
axes[1].axvline(lift_draws.mean(), color="red", lw=2, label=f"기대 향상 {lift_draws.mean():+.2f}%p")  # 세로 기준선 표시
axes[1].set_title("Posterior of lift (B - A)"); axes[1].set_xlabel("lift (%p)"); axes[1].legend()
plt.tight_layout(); plt.show()                                    # 그래프 간격 자동 정리 후 출력

print(f"향상폭이 0보다 클 확률 P(B>A) = {(dB > dA).mean():.1%}")
print(f"향상폭이 1%p보다 클 확률       = {((dB - dA) > 0.01).mean():.1%}")
print(f"손해 볼 확률 P(B<A)            = {(dB < dA).mean():.1%}")
````

> **읽는 법(그림):** 왼쪽은 두 안의 연락률 사후분포입니다. **꽤 많이 겹칩니다** — 앞의 `ethnicity` 사례와 다른 점이 이것입니다. 오른쪽은 *향상폭(B−A)* 의 분포이고, 검은 0선의 **오른쪽에 대부분(약 96%)** 이 있지만 **왼쪽에도 꼬리가 남아** 있습니다. 그 왼쪽 꼬리가 곧 "B로 바꿨는데 오히려 나빴을 확률"입니다.
>
> 💡 **여기서 두 렌즈의 성격 차이가 드러납니다.** 빈도주의는 유의수준 0.05를 기준으로 **"유의하지 않음 → 기각 실패"** 라고 이분법으로 답합니다. 베이지안은 같은 데이터에서 **"B가 나을 확률 96%, 손해 볼 확률 4%, 기대손실은 0.01%p"** 라고 답합니다. 어느 쪽이 틀린 게 아니라, **묻는 질문이 다릅니다.**
>
> 그래서 실무 결정은 이렇게 갑니다 — *변경 비용이 거의 없다면* 기대손실이 매우 작으니 채택할 만하고, *비용이 크거나 되돌리기 어렵다면* 표본을 더 모아 확신을 높이는 편이 낫습니다. **숫자 하나가 결정을 대신하지 않습니다.**

## 📋 A/B 의사결정 보고서 ("채택할까?")

위 결과로 아래 보고서를 채웁니다. **가정과 결론을 분리**하는 게 핵심입니다.

```markdown
# 이력서 품질 A/B 의사결정 보고서

## 1. 실험 설계
- 가설: 품질을 보강한 이력서(B)가 면접 연락률을 높인다
- 배정: 무작위 배정 (그룹당 약 2,4__건)
- 배정 점검(SRM): p = __ → 정상/이상

## 2. 결과 (두 렌즈)
- 관측 연락률: A __%, B __%  (절대 차이 __%p)
- 빈도주의: z = __, p = __, 95% CI [__, __]%p → 유의/비유의
- 베이지안: P(B>A) = __%, 기대 향상 __%p, 손해 볼 확률 __%, 기대손실 __%p

## 3. 의사결정 제안  ← 확률·위험·비용을 엮어서
- (예) "p는 0.05를 넘어 '유의'는 아니지만, B가 나을 확률이 __%이고 기대손실이 __%p로 매우 작다.
  변경 비용이 낮으므로 B 채택을 권고" 또는 "결정을 미루고 표본을 __건 더 확보"
- 필요 표본 추정: 관측된 차이(__%p)를 80% 검정력으로 잡으려면 그룹당 약 __건 (Part 5 방법)

## 4. 가정과 한계 ← 정직함
- 무작위 배정·SRM 정상 가정. 결과는 이 표본·이 시기·이 직군에 대한 것.
- 통계적 유의 ≠ 비즈니스 가치. 변경 비용·운영 부담을 함께 고려.
- 이 실험은 '연락 여부'만 측정 — 최종 채용·성과 같은 하류 지표는 알 수 없음.
```

> 💡 빈칸(`__`)은 위 출력으로 채우고, **3번 의사결정은 "확률·위험·비용"을 엮어** 한 문단으로 쓰세요. "p<0.05라서 채택"이 아니라 "이만한 확률과 위험이면 이 비용을 감수할 만하다"가 좋은 의사결정 언어입니다.

보고서를 쓸 때 한 가지 더 — **4번 '가정과 한계'를 절대 빈칸으로 두지 마세요.** 초보 분석가일수록 결과(2번)에 힘을 주고 한계는 대충 넘기지만, 노련한 분석가일수록 한계를 정직하게 적습니다. "노벨티 효과(새것이라 잠깐 반짝하는 현상)가 섞였을 수 있다", "실험 기간이 짧아 계절성을 반영 못 했다", "통계적으로 유의하지만 상승폭이 작아 운영 비용 대비 이득이 불확실하다" 같은 문장이 그 예입니다. 한계를 밝히는 건 분석을 깎아내리는 게 아니라, *어디까지 믿어도 되는지*를 알려주는 신뢰의 표시입니다. 의사결정자는 "무조건 확실합니다"라는 분석가보다 "이 조건 하에서 이만큼 말할 수 있습니다"라는 분석가를 더 신뢰합니다. 그게 데이터로 사람을 정직하게 설득하는 방법입니다.

## 한 번의 실험은 흔들려도, 무작위는 평균적으로 옳다

여기서 중요한 통찰 하나. _한 번의_ A/B 실험 결과는 우연히 진짜 효과에서 빗나갈 수 있습니다(표본 추출의 운). 그런데 무작위 배정의 위대함은 — 같은 실험을 *수없이 반복*하면, 그 추정치들이 **진짜 효과를 중심으로** 흩어진다는 데 있습니다. 즉 무작위 추정은 *비편향(Unbiased)*입니다. 편향(체계적으로 한쪽으로 치우침)이 없고, 흩어짐(분산)만 있습니다. 시뮬레이션으로 눈에 보이게 확인해봅시다.

````python
# [C28] 무작위화의 비편향성 — 같은 실험을 500번 반복하면?
np.random.seed(5)                                                 # 재현성: 난수를 항상 같게 고정
true_effect = 5.0
estimates = []
for _ in range(500):
    X = np.random.normal(0, 1, 1500)                 # 교란
    T = np.random.binomial(1, 0.5, 1500)             # 무작위 배정
    Y = 50 + true_effect*T + 12*X + np.random.normal(0, 5, 1500)
    estimates.append(Y[T == 1].mean() - Y[T == 0].mean())
estimates = np.array(estimates)

fig, ax = plt.subplots(figsize=(8, 4))                            # 그래프 하나를 그릴 도화지 준비
ax.hist(estimates, bins=30, color="steelblue", alpha=0.7)
ax.axvline(true_effect, color="indianred", lw=2, ls="--", label=f"true effect = {true_effect}")  # 세로 기준선 표시
ax.axvline(estimates.mean(), color="black", lw=1.5, label=f"mean of estimates = {estimates.mean():.2f}")  # 세로 기준선 표시
ax.set_xlabel("ATE estimate (per experiment)"); ax.set_ylabel("count")
ax.set_title("500 randomized experiments: estimates center on the truth"); ax.legend()
plt.tight_layout(); plt.show()                                    # 그래프 간격 자동 정리 후 출력
print(f"500개 추정치의 평균 = {estimates.mean():.2f} (진짜 효과 {true_effect}에 수렴) → 비편향")
````

> **읽는 법(그림):** 개별 실험(막대)은 진짜 효과(빨간 선) 주위로 흩어지지만, 그 _평균_(검은 선)은 진짜 효과에 정확히 포개집니다. 한 번의 실험이 빗나갈 수 있어도 _체계적 편향은 없다_ — 이것이 무작위화가 인과 추정의 황금 표준인 이유입니다. (Part 4의 관찰 데이터였다면 이 분포가 진짜 효과에서 _치우쳐_ 있었을 겁니다.)

## A/B는 비율만의 도구가 아니다 — 연속형 결과의 무작위 실험

지금까지는 결과가 "연락 옴 / 안 옴"이라는 **비율**이었습니다. 그런데 실무의 관심 지표는 *체류 시간·매출액·점수*처럼 **연속형**인 경우도 많습니다. 그때는 도구만 바뀝니다 — 비율 z-검정 대신 **2표본 t-test**(D013!)를 쓰고, 효과는 평균 차이와 Cohen's d로 보고합니다.

**Project STAR**(`star`)로 확인해봅시다. 테네시주에서 학생을 학급 유형에 *무작위 배정*한 실제 대규모 실험입니다. 여기서 ATE는 곧 "소규모 학급의 평균 점수 − 일반 학급의 평균 점수"입니다.

````python
# [C28b] 연속형 결과의 무작위 실험 — ATE = 두 집단 평균의 차이
small_c = star.loc[star["classk"] == "small.class", "tmathssk"]
regular = star.loc[star["classk"] == "regular",     "tmathssk"]

ate = small_c.mean() - regular.mean()
t_st, p_st = stats.ttest_ind(small_c, regular, equal_var=False)      # Welch
se = np.sqrt(small_c.var(ddof=1) / len(small_c) + regular.var(ddof=1) / len(regular))
lo_st, hi_st = ate - 1.96 * se, ate + 1.96 * se
sp = np.sqrt(((len(small_c) - 1) * small_c.var(ddof=1) + (len(regular) - 1) * regular.var(ddof=1))
             / (len(small_c) + len(regular) - 2))

print("[Project STAR — 소규모 학급 vs 일반 학급 (유치원 수학 점수)]")
print(f"  n = {len(small_c)} vs {len(regular)}")
print(f"  평균 {small_c.mean():.1f} vs {regular.mean():.1f}")
print(f"  ① ATE(평균 차이) : {ate:+.2f}점")
print(f"  ② 95% 신뢰구간   : [{lo_st:+.2f}, {hi_st:+.2f}]점")
print(f"  ③ 효과 크기 d    : {ate/sp:.3f}")
print(f"  (참고: Welch t = {t_st:.2f}, p = {p_st:.2e})")
````

> **읽는 법:** 무작위 배정이므로 두 집단 평균의 차이가 그대로 **ATE 추정치**입니다 — 소규모 학급이 약 8점 높고, 신뢰구간이 0을 포함하지 않습니다. *분석 도구는 D013의 2표본 t-test와 완전히 같습니다.* 달라진 것은 **데이터가 만들어진 방식**(무작위 배정)이고, 그 덕분에 우리는 이 차이를 "연관"이 아니라 **"학급 규모의 효과"** 로 부를 수 있습니다.
>
> 다만 효과 크기 d는 0.17로 **작습니다.** "무작위 실험으로 확인된 진짜 효과"와 "큰 효과"는 다른 이야기라는 점 — 인과성과 효과 크기는 별개의 축입니다. 정책 결정은 이 8점이 *학급 축소 비용*을 정당화하는지까지 따져야 합니다.
>
> 💡 **정리하면:** A/B 테스트의 본질은 *분석 기법*이 아니라 **무작위 배정이라는 설계**입니다. 결과가 비율이면 z-검정·카이제곱, 연속형이면 t-test, 여러 집단이면 ANOVA — 도구는 지난 시간들에 이미 다 배웠습니다.

| ----------------------------------- | ----------------------------------------------------------- |
| 무작위 실험을 _할 수 있다_          | **A/B 테스트** — 이 노트북에서 다룬 내용                    |
| 실험이 *불가능*하다 (관찰 데이터뿐) | 준실험 설계 — 이중차분(DiD)·성향점수매칭(PSM)·회귀단절(RDD) |
| 효과가 사람마다 다른지 알고 싶다    | 이질적 처치효과(HTE) 추정 — 인과 머신러닝 계열              |

> 📌 **이 노트북의 범위:** 여기서는 **무작위 실험이 가능한 경우**에 집중했습니다. 무작위 배정이 있었기 때문에 "처치군 평균 − 대조군 평균"을 그대로 인과 효과로 읽을 수 있었죠. 반대로 실험이 불가능한 상황(이미 시행된 정책, 과거 데이터)에서는 위 표의 준실험 설계가 필요합니다. 그 방법들도 결국 **"비교 가능한 대조군을 어떻게 구하나"**라는 오늘의 질문에서 출발하니, 여기서 익힌 사고가 그대로 이어집니다.

> 💡 **한 문장으로:** 인과 추정의 승부처는 분석 기법이 아니라 **설계와 가정**입니다. "이 두 집단이 정말 비교 가능한가?"에 자신 있게 답할 수 있을 때, 비로소 어떤 분석을 얹어도 신뢰할 수 있습니다.

# 📝 오늘의 과제

## 과제 — A/B 실험 분석 보고서 (설계 → 두 렌즈 → 의사결정)

### 1. 데이터 — 라이브러리가 제공하는 **무작위 실험** 데이터에서 고르기

인과 주장을 하려면 데이터가 *무작위 배정*으로 만들어졌어야 합니다. 아래는 모두 실제 무작위 실험(또는 실험 자료)입니다.

```python
import statsmodels.api as sm

# (a) 이력서 현장실험 — 이름·품질이 무작위 배정, 결과=연락 여부(비율)
resume = sm.datasets.get_rdataset("ResumeNames", "AER").data
#   추가 분석 아이디어: gender × call, city(보스턴/시카고)별로 나눠 비교(서브그룹 분석)

# (b) Project STAR — 학급 유형이 무작위 배정, 결과=시험 점수(연속형)
star = sm.datasets.get_rdataset("Star", "Ecdat").data
#   또는 더 풍부한 버전: sm.datasets.get_rdataset("STAR", "AER").data

# (c) RAND 건강보험 실험 — 보험 급여율이 무작위 배정, 결과=의료 이용
randhie = sm.datasets.randhie.load_pandas().data      # statsmodels 내장
```

> 💡 **비교용으로 관찰 데이터를 하나 더 골라도 좋습니다.** 같은 질문을 무작위 실험 데이터와 관찰 데이터에서 각각 분석해, 결론이 어떻게 달라지는지 비교하면 오늘의 교훈이 몸에 남습니다.

### 2. 분석 (아래를 모두 포함)

- [ ] **설계 점검:** 배정 비율(SRM) 점검, 표본 크기와 **MDE** 계산 — "이 표본으로 어느 크기의 효과까지 잡을 수 있나"
- [ ] **빈도주의 분석:** 결과가 비율이면 두 비율 z-검정, 연속형이면 2표본 t-test. **점추정 + 95% CI + p**를 함께 보고
- [ ] **베이지안 분석:** (비율일 때) 베타 사후분포, **P(B>A)**, 기대 향상, **기대손실**
- [ ] **효과 크기:** 절대 차이와 상대 차이(또는 Cohen's d)를 함께 — 상대 차이만 쓰지 않기
- [ ] **의사결정 보고서:** 위 양식으로, *가정·결론·한계를 분리*해서 작성

### 3. 심화 (선택)

- [ ] **서브그룹 분석**을 해보고, 그것이 왜 위험한지(다중비교!) 함께 적기
- [ ] peeking 시뮬레이션을 직접 돌려, 여러 번 들여다볼 때 거짓 양성률이 어떻게 변하는지 보이기
- [ ] 관찰 데이터에서 같은 비교를 해보고 **무작위 실험 결과와 얼마나 다른지** 비교 (D017의 예고편이 됩니다)
